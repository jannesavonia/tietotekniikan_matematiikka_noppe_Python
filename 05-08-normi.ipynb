    $\mathbb{R}^n$ vektoreille.

<hr class="theorem-block-start" style="border:0; border-top:4px solid #1565c0; margin-top:1.2em; margin-bottom:0.6em;">

#### Definition

Vektorin $\textbf{x}$ pituus, eli normi
$$\lVert \mathbf{x} \rVert = \sqrt{x_1^2 + x_2^2 + \dots + x_n^2}\ .$$

<hr class="theorem-block-end" style="border:0; border-top:2px solid #1565c0; margin-top:0.6em; margin-bottom:1.2em;">

<hr class="theorem-block-start" style="border:0; border-top:4px solid #00695c; margin-top:1.2em; margin-bottom:0.6em;">

#### Remark

Yleisesti ei päde:

1.  $\lVert \mathbf{x} \rVert = \lVert \mathbf{y} \rVert \Rightarrow \mathbf{x}=\mathbf{y}$

2.  $\lVert \mathbf{x} \rVert+\lVert \mathbf{y} \rVert= \lVert \mathbf{x}+\mathbf{y} \rVert$

<hr class="theorem-block-end" style="border:0; border-top:2px solid #00695c; margin-top:0.6em; margin-bottom:1.2em;">

<hr class="theorem-block-start" style="border:0; border-top:4px solid #6a1b9a; margin-top:1.2em; margin-bottom:0.6em;">

#### Theorem

Olkoon $\mathbf{x}$ vektori ja $a \in \mathbb{R}$. Tällöin
$$\lVert a \mathbf{x} \rVert = \lvert a \rvert \lVert \mathbf{x} \rVert\ .$$

<hr class="theorem-block-end" style="border:0; border-top:2px solid #6a1b9a; margin-top:0.6em; margin-bottom:1.2em;">

<hr class="theorem-block-start" style="border:0; border-top:4px solid #ef6c00; margin-top:1.2em; margin-bottom:0.6em;">

#### Example 22

$$
\begin{aligned}
            \lVert \begin{pmatrix}1& 2\end{pmatrix} \rVert = & \sqrt{1^2 + 2^2} = \sqrt{5} \\
            \lVert \begin{pmatrix}-1& -2\end{pmatrix} \rVert = & \sqrt{(-1)^2 + (-2)^2} = \sqrt{5} \\
            \lVert \begin{pmatrix}2& -1 3\end{pmatrix} \rVert & = & \sqrt{2^2 + (-1)^2 + 3^2} = \sqrt{14} \\
            \lVert 3 \begin{pmatrix}1& 2\end{pmatrix} \rVert = & \lvert 3 \rvert \lVert \begin{pmatrix}1& 2\end{pmatrix} \rVert  = 3 \sqrt{5} \\
            \lVert \begin{pmatrix}1& 2\end{pmatrix} + \begin{pmatrix}1 1\end{pmatrix} \rVert & = &  \lVert \begin{pmatrix}2& 3\end{pmatrix} \rVert  = \underbrace{\sqrt{13}}_{\approx 3.61} \\
            & \le \lVert \begin{pmatrix}1& 2\end{pmatrix} \rVert + \lVert \begin{pmatrix}1 & 1\end{pmatrix} \rVert = \underbrace{\sqrt{5} + \sqrt{2}}_{\approx 3.65}
\end{aligned}
$$
