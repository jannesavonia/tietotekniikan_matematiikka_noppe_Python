# Salaukset

## Transpositiosalaus

### Transpositiosalaus

-   [Transpositiosalaus](https://fi.wikipedia.org/wiki/Transpositiosalaus)

-   Ideana vaihtaa kirjainten paikkoja avaimen mukaan

### Salaus

1.  Olkoon $K=(3,\, 6,\, 7,\, 4,\, 2,\, 1,\, 5)$, avaimen pituus $l=7$

2.  Olkoon viesti  
    $M$=`"TAMA VIESTI ON SALATTU TRANSPOSITIOSALAUKSELLA"`

3.  Jaetaan viesti $l=7$ sarakkeeseen:  

    |  3  |  6  |  7  |  4  |  2  |  1  |  5  |
    |:---:|:---:|:---:|:---:|:---:|:---:|:---:|
    |  T  |  A  |  M  |  A  |     |  V  |  I  |
    |  E  |  S  |  T  |  I  |     |  O  |  N  |
    |     |  S  |  A  |  L  |  A  |  T  |  T  |
    |  U  |     |  T  |  R  |  A  |  N  |  S  |
    |  P  |  O  |  S  |  I  |  T  |  I  |  O  |
    |  S  |  A  |  L  |  A  |  U  |  K  |  S  |
    |  E  |  L  |  L  |  A  |     |     |     |

1.  Permutoidaan sarakkeet avaimen mukaiseen järjestykseen:  

    |  1  |  2  |  3  |  4  |  5  |  6  |  7  |
    |:---:|:---:|:---:|:---:|:---:|:---:|:---:|
    |  V  |     |  T  |  A  |  I  |  A  |  M  |
    |  O  |     |  E  |  I  |  N  |  S  |  T  |
    |  T  |  A  |     |  L  |  T  |  S  |  A  |
    |  N  |  A  |  U  |  R  |  S  |     |  T  |
    |  I  |  T  |  P  |  I  |  O  |  O  |  S  |
    |  K  |  U  |  S  |  A  |  S  |  A  |  L  |
    |     |     |  E  |  A  |     |  L  |  L  |

2.  Kryptattu viesti  
    $C$=`"V TAIAMO EINSTTA LTSANAURS TITPIOOSKUSASAL  EA LL"`

### Purkaminen

1.  $K=(3,\, 6,\, 7,\, 4,\, 2,\, 1,\, 5)$

2.  Jaetaan viesti $l=7$ sarakkeeseen:  

    |  3  |  6  |  7  |  4  |  2  |  1  |  5  |
    |:---:|:---:|:---:|:---:|:---:|:---:|:---:|
    |  1  |  2  |  3  |  4  |  5  |  6  |  7  |
    |  V  |     |  T  |  A  |  I  |  A  |  M  |
    |  O  |     |  E  |  I  |  N  |  S  |  T  |
    |  T  |  A  |     |  L  |  T  |  S  |  A  |
    |  N  |  A  |  U  |  R  |  S  |     |  T  |

    |  K  |  U  |  S  |  A  |  S  |  A  |  L  |

    saman numeron alle avaimessa (ensimmäinen rivi).

    |  3  |  6  |  7  |  4  |  2  |  1  |  5  |

    |     |  S  |  A  |  L  |  A  |  T  |  T  |
    |  U  |     |  T  |  R  |  A  |  N  |  S  |
    |  P  |  O  |  S  |  I  |  T  |  I  |  O  |
    |  S  |  A  |  L  |  A  |  U  |  K  |  S  |

2.  $M=$`"TAMA VIESTI ON SALATTU TRANSPOSITIOSALAUKSELLA  "`
