# Matriisit

## Determinantti

<hr class="theorem-block-start" style="border:0; border-top:4px solid #00695c; margin-top:1.2em; margin-bottom:0.6em;">

#### Remark

Käänteismatriisin määritelmä ei antanut juurikaan työkaluja
käänteismatriisin muodostamiseksi, tai edes sen olemassaolon
selvittämiseksi. Tähän ongelmaan ratkaisun tarjoaa matriisin
determinantti.

<hr class="theorem-block-end" style="border:0; border-top:2px solid #00695c; margin-top:0.6em; margin-bottom:1.2em;">

<hr class="theorem-block-start" style="border:0; border-top:4px solid #1565c0; margin-top:1.2em; margin-bottom:0.6em;">

#### Definition

Olkoon $\textbf{A} \in \mathbb{R}^{n \times n}$. Määritellään $\mathbf{A}$:n
**determinantti** rekursiivisesti
$$| \mathbf{A} | =\rm{det} \mathbf{A} = \sum_{j=1}^{n} (-1)^{1+j} a_{1,j} \rm{det} \mathbf{A}^{\{1,j\}} \ ,$$

Tässä määritelty determinantti on kehitetty 1. rivin suhteen

> *Määritelmä voidaan antaa yleisemminkin, mutta tämä on riittävä
> määritelmä tässä yhteydessä.*

. Lisäksi $\mathbf{A}^{\{1, j\}}$ on sellainen $\textbf{A}$:n alimatriisi,
josta on poistettu 1. rivi ja $j$:s sarake.  
Määräritellään lisäksi, että $1 \times 1$ matriisin
$\begin{pmatrix}a\end{pmatrix}$ determinantti on $a$.

<hr class="theorem-block-end" style="border:0; border-top:2px solid #1565c0; margin-top:0.6em; margin-bottom:1.2em;">

<hr class="theorem-block-start" style="border:0; border-top:4px solid #ef6c00; margin-top:1.2em; margin-bottom:0.6em;">

#### Example 49

Olkoon 

$$
\mathbf{A} = \begin{pmatrix}
      3 & 5\\
      1 & 2
    \end{pmatrix} .
$$

 Tällöin 

$$
\begin{aligned}
    \rm{det}  \mathbf{A}  & = 3\, \rm{det}(2) - 5\, \rm{det}(1) \\
    &= 3 \cdot 2 - 5 \cdot 1 = 1\ .
\end{aligned}
$$

<hr class="theorem-block-end" style="border:0; border-top:2px solid #ef6c00; margin-top:0.6em; margin-bottom:1.2em;">

<hr class="theorem-block-start" style="border:0; border-top:4px solid #ef6c00; margin-top:1.2em; margin-bottom:0.6em;">

#### Example 50

Olkoon $\textbf{A} \in \mathbb{R}^{2 \times 2}$. Tällöin sen
determinantti 

$$
\begin{aligned}
        \rm{det} \mathbf{A}  & = \sum_{j=1}^{n} (-1)^{i+j} a_{ij}  \rm{det} \mathbf{A}^{\{ij\}} \\
        & = (-1)^{1+1} a_{1,1} a_{2,2} + (-1)^{1+2} a_{1,2} a_{2,1}\\
        & = a_{1,1} a_{2,2} - a_{1,2} a_{2,1}
\end{aligned}
$$

<hr class="theorem-block-end" style="border:0; border-top:2px solid #ef6c00; margin-top:0.6em; margin-bottom:1.2em;">

<hr class="theorem-block-start" style="border:0; border-top:4px solid #ef6c00; margin-top:1.2em; margin-bottom:0.6em;">

#### Example 51

Olkoon $\mathbf{A}=\begin{pmatrix}
      2 & -1 & 3\\
      0 & 4  & 5\\
      1 & 2  & -2
    \end{pmatrix}$. Tällöin 

$$
\begin{aligned}
      \rm{det} \mathbf{A}  & = 2\, \rm{det} \begin{pmatrix}4 & 5\\ 2 & -2\end{pmatrix}
      -(-1)\, \rm{det} \begin{pmatrix}0 & 5\\ 1 -2\end{pmatrix}
      +3\, \rm{det} \begin{pmatrix}0 & 4\\ 1 2\end{pmatrix}\\
      & = \underbrace{2\,(4\cdot(-2)-5\cdot 2)}_{=-36}-\underbrace{(-1)\,(0\cdot(-2)-5\cdot 1)}_{=5}+\underbrace{3\,(0\cdot 2 - 4\cdot 1)}_{=-12}\\
      &= -53\ .
\end{aligned}
$$

#### Example 52

Olkoon $\mathbf{A} \in \mathbb{R}^{3 \times 3}$. Tällöin sen determinantti

$$
\begin{aligned}
        \rm{det} \mathbf{A}  & = \sum_{j=1}^{n} (-1)^{i+j} a_{ij}\, \rm{det} \mathbf{A}^{\{ij\}} \\
        & = a_{1,1} \rm{det} \mathbf{A}^{\{1, 1\}} - a_{1,2} \rm{det} \mathbf{A}^{\{1, 2\}} + a_{1,3} \rm{det} \mathbf{A}^{\{1, 3\}} \\
        & = a_{1,1} (a_{2,2} a_{3,3} - a_{3,2} a_{2,3}) - a_{1,2} (a_{2,1} a_{3,3} - a_{3,1} a_{2,3})\\
        & + a_{1,3} (a_{2,1} a_{3,2} - a_{3,1} a_{2,2}) \\
\end{aligned}
$$

#### Theorem

Olkoon $\mathbf{A},\,\mathbf{B} \in \mathbb{R}^{n \times n}$. Tällöin

1.  $\rm{det} \mathbf{A} \neq 0 \Leftrightarrow \textbf{A}$ on kääntyvä.

2.  jos $\mathbf{A}$ on kääntyvä, niin
    $\rm{det} \mathbf{A}^{-1} = \frac{1}{\rm{det} \mathbf{A}}$.

3.  jos jokin $\mathbf{A}$:n sarake tai rivi koostuu pelkästään nollista,
    niin $\rm{det} \mathbf{A}=0$.

4.  $\rm{det} (\mathbf{AB})  =  \rm{det} \mathbf{A}\, \rm{det} \mathbf{B}$.

5.  $\rm{det} (\mathbf{A}^T)  =  \rm{det} \mathbf{A}$.

6.  $\rm{det} \mathbf{I}  =  1$.
