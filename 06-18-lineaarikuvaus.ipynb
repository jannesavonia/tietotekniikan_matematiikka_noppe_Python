# Matriisit

## Lineaarikuvaus

<hr class="theorem-block-start" style="border:0; border-top:4px solid #1565c0; margin-top:1.2em; margin-bottom:0.6em;">

#### Definition

Kuvaus $\mathbf{A}:\mathbb{R}^m\rightarrow\mathbb{R}^n$ on **lineaarinen**, jos

1.  $\mathbf{A}(c \mathbf{x}) = c\mathbf{A}(\mathbf{x})$ ja

2.  $\mathbf{A}(\mathbf{x}+\mathbf{y}) = \mathbf{A}(\mathbf{x}) + \mathbf{A}(\mathbf{y})$

kaikilla $c \in \mathbb{R}$, $\mathbf{x},\mathbf{y} \in \mathbb{R}^m$.

<hr class="theorem-block-end" style="border:0; border-top:2px solid #1565c0; margin-top:0.6em; margin-bottom:1.2em;">

-   Lineaarikuvaukset säilyttävät suorat suorina

-   Lineaarikuvaus kuvaa origon origoksi

-   Kierrot

-   Skaalaukset

-   Venytykset

-   Peilaukset

-   Kaikki näiden yhdistelmät

<hr class="theorem-block-start" style="border:0; border-top:4px solid #ef6c00; margin-top:1.2em; margin-bottom:0.6em;">

#### Example 54

$\mathbf{A}(\mathbf{x}) = \begin{pmatrix}1 & 2 \\ 0 & 1 \\ 1 & 0\end{pmatrix} \mathbf{x}$ on
lineaarikuvaus $\mathbb{R}^2 \rightarrow \mathbb{R}^3$, koska 

$$
\begin{aligned}
        c \mathbf{A}(\mathbf{x}) & = c \begin{pmatrix}1 & 2 \\ 0 & 1 \\ 1 & 0\end{pmatrix} \begin{pmatrix}x_1\\x_2\end{pmatrix}
        =  c \begin{pmatrix}1 \cdot x_1 + 2 \cdot x_2 \\ 0 \cdot x_1 + 1 \cdot x_2 \\ 1 \cdot x_1 +  0 \cdot x_2 \end{pmatrix}\\
        & = \begin{pmatrix}1 \cdot c\cdot x_1 + 2 \cdot c \cdot x_2 \\ 0 \cdot c\cdot x_1 + 1 \cdot c\cdot  x_2 \\ 1 \cdot c\cdot  x_1 +  0 \cdot c\cdot x_2 \end{pmatrix}  =\begin{pmatrix}1 & 2 \\ 0 & 1 \\ 1 & 0\end{pmatrix} \begin{pmatrix}cx_1\\cx_2\end{pmatrix} \\
        & = \mathbf{A}(c \mathbf{x}) \\[1.5ex]
        \mathbf{A}(\mathbf{x}+ \mathbf{y}) & = \begin{pmatrix}1 & 2 \\ 0 & 1 \\ 1 & 0\end{pmatrix} (\mathbf{x} + \mathbf{y})  = \begin{pmatrix}1 & 2 \\ 0 & 1 \\ 1 & 0\end{pmatrix} \mathbf{x} + \begin{pmatrix}1 & 2 \\ 0 & 1 \\ 1 & 0\end{pmatrix} \mathbf{y} \\
        & = \mathbf{A}(\mathbf{x}) + \mathbf{A}(\mathbf{y})\ .
\end{aligned}
$$

#### Example 55

$\mathbf{B}(\mathbf{x}) = \mathbf{x} + \begin{pmatrix}1 \\ 0\end{pmatrix}$ ei ole lineaarikuvaus, koska

$$
\begin{aligned}
        \mathbf{B}(\begin{pmatrix}0 \\ 1\end{pmatrix} + \begin{pmatrix}1 \\ 1\end{pmatrix}) & = \begin{pmatrix}1 \\ 2\end{pmatrix} + \begin{pmatrix}1 \\ 0\end{pmatrix}  = \begin{pmatrix}2 \\ 2\end{pmatrix}   \neq  \begin{pmatrix}3 \\ 2\end{pmatrix}\\
        & = \begin{pmatrix}1 \\ 1\end{pmatrix} + \begin{pmatrix}2\\ 1\end{pmatrix}  =  \begin{pmatrix}0 \\ 1\end{pmatrix} + \begin{pmatrix}1 \\ 0\end{pmatrix} + \begin{pmatrix}1 \\ 1\end{pmatrix} + \begin{pmatrix}1 \\ 0\end{pmatrix}\\
        & = \mathbf{B}(\begin{pmatrix}0 \\ 1\end{pmatrix}) + \mathbf{B}(\begin{pmatrix}1 \\ 1\end{pmatrix})\ .
\end{aligned}
$$

<hr class="theorem-block-end" style="border:0; border-top:2px solid #ef6c00; margin-top:0.6em; margin-bottom:1.2em;">

#### Theorem

Jokainen lineaarikuvaus $\mathbb{R}^m \rightarrow \mathbb{R}^n$ voidaan esittää
$\mathbb{R}^{n \times m}$ matriisin ja kertolaskun avulla.  
Vastaavasti jokainen matriisi $\mathbb{R}^{n \times m}$ voidaan samaistaa
lineaarikuvaukseen $\mathbb{R}^m \rightarrow \mathbb{R}^n$.

<hr class="theorem-block-end" style="border:0; border-top:2px solid #6a1b9a; margin-top:0.6em; margin-bottom:1.2em;">

<hr class="theorem-block-start" style="border:0; border-top:4px solid #00695c; margin-top:1.2em; margin-bottom:0.6em;">

#### Remark

Laskuteknisesti matriisit ja lineaarikuvaukset ovat siis sama asia.

<hr class="theorem-block-end" style="border:0; border-top:2px solid #00695c; margin-top:0.6em; margin-bottom:1.2em;">

<hr class="theorem-block-start" style="border:0; border-top:4px solid #00695c; margin-top:1.2em; margin-bottom:0.6em;">

#### Remark

Olkoon $\mathbf{A}(\mathbf{x})$ lineaarikuvaus ja $\mathbf{A}$ sitä vastaava
matriisit. Tällöin $$\mathbf{A}(\mathbf{x}) = \mathbf{A}\mathbf{x}.$$

<hr class="theorem-block-end" style="border:0; border-top:2px solid #00695c; margin-top:0.6em; margin-bottom:1.2em;">

<hr class="theorem-block-start" style="border:0; border-top:4px solid #6a1b9a; margin-top:1.2em; margin-bottom:0.6em;">

#### Theorem

Olkoon $\mathbf{A}(\mathbf{x})$ ja $\mathbf{B}(\mathbf{y})$ lineaarikuvauksia ja $\mathbf{A}$
ja $\mathbf{B}$ niitä vastaavat matriisit. Tällöin
$$\mathbf{B}\Big(\mathbf{A}(\mathbf{x})\Big) = \mathbf{B}\mathbf{A}\mathbf{x}.$$
