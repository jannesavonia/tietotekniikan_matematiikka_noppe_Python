# Vektorit

## Yksikkövektori ja samansuuntaisuus

<hr class="theorem-block-start" style="border:0; border-top:4px solid #1565c0; margin-top:1.2em; margin-bottom:0.6em;">

#### Definition

**Yksikkövektori** on vektori jonka pituus on $1$.

<hr class="theorem-block-end" style="border:0; border-top:2px solid #1565c0; margin-top:0.6em; margin-bottom:1.2em;">

<hr class="theorem-block-start" style="border:0; border-top:4px solid #1565c0; margin-top:1.2em; margin-bottom:0.6em;">

#### Definition

1.  Olkoon $\mathbf{x}, \mathbf{y} \in \mathbb{R}^n$. Tällöin $\mathbf{x}$ on
    **samansuuntainen** kuin $\mathbf{y}$ jos on olemassa $a > 0$ s.e
    $\mathbf{x}=a\mathbf{y}$.

2.  Jos $-\mathbf{x}$ on samansuuntainen kuin $\mathbf{y}$, niin tällöin
    vektorit ovat **vastakkaissuuntaiset**.

3.  Jos vektorit ovat samansuuntaisia tai vastakkaissuuntaisia, niin
    tällöin vektorit ovat **yhdensuuntaiset**.

<hr class="theorem-block-end" style="border:0; border-top:2px solid #1565c0; margin-top:0.6em; margin-bottom:1.2em;">

<hr class="theorem-block-start" style="border:0; border-top:4px solid #6a1b9a; margin-top:1.2em; margin-bottom:0.6em;">

#### Theorem

Olkoon $\mathbf{x} \in \mathbb{R}^n$, $\lVert \mathbf{x} \rVert \ne 0$. Tällöin $\mathbf{x}$:n
kanssa samansuuntainen yksikkövektori
$\hat{\mathbf{x}} = \frac{1}{\lVert \mathbf{x} \rVert} \mathbf{x}$.

<hr class="theorem-block-end" style="border:0; border-top:2px solid #6a1b9a; margin-top:0.6em; margin-bottom:1.2em;">

<hr class="theorem-block-start" style="border:0; border-top:4px solid #00695c; margin-top:1.2em; margin-bottom:0.6em;">

#### Remark

1.  Usein käytetään termiä $\mathbf{x}$:n suuntainen yksikkövektori.

2.  Jos $\mathbf{x} \in \mathbb{R}^n$, niin jatkossa $\hat{\mathbf{x}}$ on $\mathbf{x}$:n
    suuntainen yksikkövektori.

<hr class="theorem-block-end" style="border:0; border-top:2px solid #00695c; margin-top:0.6em; margin-bottom:1.2em;">

<hr class="theorem-block-start" style="border:0; border-top:4px solid #ef6c00; margin-top:1.2em; margin-bottom:0.6em;">

#### Example 24

Olkoon $\mathbf{x}=\begin{pmatrix}9 & 2 & -7\end{pmatrix}$. Tällöin 

$$
\begin{aligned}
            \lVert \mathbf{x} \rVert & = \sqrt{9^2 + 2^2 + (-7)^2}\\
      & = \sqrt{134} \\
            \hat{\mathbf{x}} & = \frac{1}{\sqrt{134}} \mathbf{x} \ .
\end{aligned}
$$

<hr class="theorem-block-end" style="border:0; border-top:2px solid #ef6c00; margin-top:0.6em; margin-bottom:1.2em;">

<hr class="theorem-block-start" style="border:0; border-top:4px solid #ef6c00; margin-top:1.2em; margin-bottom:0.6em;">

#### Example 25

Tiedetään, että
$\hat{\mathbf{x}}=\begin{pmatrix}\sqrt{\frac{1}{2}} & -\sqrt{\frac{1}{2}}\end{pmatrix}$ ja
$\lVert \mathbf{x} \rVert=\sqrt{8}$, lasketaan $\mathbf{x}$.  

1.  Tiedetään, että $\hat{\mathbf{x}} = \frac{1}{ \lVert \mathbf{x} \rVert}\mathbf{x}$.

2.  Siten 

$$
\begin{aligned}
                    \mathbf{x}& = \lVert \mathbf{x} \rVert\hat{\mathbf{x}} \\
                    & = \sqrt{8} \begin{pmatrix}\sqrt{\frac{1}{2}} & -\sqrt{\frac{1}{2}}\end{pmatrix} \\
                    & = \begin{pmatrix}\sqrt{8}\sqrt{\frac{1}{2}} & -\sqrt{8}\sqrt{\frac{1}{2}}\end{pmatrix} \\
                    & = \begin{pmatrix}\sqrt{4} & -\sqrt{4}\end{pmatrix} \\
                    & = \begin{pmatrix}2 & -2\end{pmatrix} \\
    \end{aligned}
$$
