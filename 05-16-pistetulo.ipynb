# Vektorit

## Pistetulo

<hr class="theorem-block-start" style="border:0; border-top:4px solid #1565c0; margin-top:1.2em; margin-bottom:0.6em;">

#### Definition

Olkoon $\mathbf{x}, \mathbf{y} \in \mathbb{R}^n$. Tällöin **pistetulo**
(sisätulo)
$$\mathbf{x} \cdot \mathbf{y} = \sum_{i=1}^n x_i y_i\ = x_1 y_1 + x_2 y_2 + \dots + x_n y_n.$$

<hr class="theorem-block-end" style="border:0; border-top:2px solid #1565c0; margin-top:0.6em; margin-bottom:1.2em;">

<hr class="theorem-block-start" style="border:0; border-top:4px solid #ef6c00; margin-top:1.2em; margin-bottom:0.6em;">

#### Example 30

Olkoon $\mathbf{x}=\begin{pmatrix}0& 2& -1\end{pmatrix}$ ja $\mathbf{y}=\begin{pmatrix}\pi& 2& 5\end{pmatrix}$.
Tällöin 

$$
\begin{aligned}
            \mathbf{x}\cdot\mathbf{y} & = \sum_{i=1}^{3}x_i y_i \\
            & = 0 \cdot \pi + 2 \cdot 2 + (-1) \cdot 5 \\
            & = 0 + 4 -5 = -1
\end{aligned}
$$

<hr class="theorem-block-end" style="border:0; border-top:2px solid #ef6c00; margin-top:0.6em; margin-bottom:1.2em;">

#### Theorem

Olkoon $\mathbf{x}, \mathbf{y}$ ja $\mathbf{z} \in \mathbb{R}^n$ sekä $a \in \mathbb{R}$.
Tällöin 

$$
\begin{aligned}
            \mathbf{x} \cdot \mathbf{y} & = \mathbf{y} \cdot \mathbf{x} \ , \\
            \mathbf{x} \cdot (\mathbf{y} + \mathbf{z}) & = \mathbf{x} \cdot \mathbf{y} + \mathbf{x} \cdot \mathbf{z} \ ,\\
            (a\mathbf{x}) \cdot \mathbf{y} & = a(\mathbf{x} \cdot \mathbf{y}) \text{ ja}\\
            \mathbf{x} \cdot \mathbf{y} & = \lVert \mathbf{x} \rVert\lVert \mathbf{y} \rVert \cos \gamma \ ,
\end{aligned}
$$

 jossa $\gamma$ on vektoreiden $\mathbf{x}$ ja $\textbf{y}$
välinen kulma.
