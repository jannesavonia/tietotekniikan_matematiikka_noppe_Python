# Differentiaaliyhtälöiden numeerinen ratkaiseminen

## Harjoituksia

### Harjoituksia

-   **M29.** Raketin massa on $100\,\rm{kg}$ ja sen moottorin
    työntövoima on $2000\,\rm{N}$. Oletetaan, että alkunopeus ja -paikka
    ovat molemmat $0$. Raketti ammutaan suoraan ylöspäin siten, että
    moottori on aluksi päällä $1 \,\rm{s}$ ja sitten pois päältä saman
    ajan käynnistyen taas $1\,\rm{s}$ ajaksi jne. Kirjoita ohjelma, joka
    simuloi raketin lentokorkeutta ajan funktiona. Ohjelma tallentaa
    tiedostoon `output.txt` raketin korkeuden ajanhetkinä
    $t=0,\, 1,\, 2,\, \dots 10$ (sekuntia). Tiedoston ensimmäinen sarake
    sisältää ajan (sekunteina) ja toinen korkeuden (metreinä). Yksiköitä
    ei talleteta tiedostoon. Sarakkeet erotetaan välilyönnillä. Käytä
    gravitaatiokiihtyvyydelle arvoa $g=10 \,\rm{m}/\rm{s}^2$ ja jätetään
    ilmanvastus huomioimatta.

-   **M30.** Ratkaise numeerisesti
    $$P\textnormal{'}(t) = P(t)\cdot (1-P(t))\ ,$$ kun
    $P(0)=\frac{1}{2}$. Talleta ratkaisu tiedostoon `output.txt`
    ajanhetkinä $t=0,\, 1,\, 2,\, \dots 10$. Tiedoston formaatti on sama
    kuin tehtävässä (M29), paitsi toinen sarake sisältää $P$:n.  
    Numeerisen ratkaisun arvot saavat poiketa oikeasta enintään
    $0.0000001$ yksikköä.

-   **M31.** Olkoon
    $$\mathbf{a}(t) = -\frac{\lVert \mathbf{v}(t) \rVert^2}{\lVert \mathbf{x}(t) \rVert^2}\mathbf{x}(t)$$
    alkuehdolla $\mathbf{x}(0)=\begin{pmatrix}2\\0\end{pmatrix}$ ja $\mathbf{v}(0)=\begin{pmatrix}0\\2\end{pmatrix}$.
    Kirjoita funktio `solver`, joka ratkaisee tämän numeerisesti ja
    tallettaa ratkaisun tiedostoon `output.txt` ajanhetkinä
    $t=0,\, 1,\, 2,\, \dots 20$ käyttäen `save(outf, x, t)`-funktiota.
    Tässä

    -   on tiedosto-olio,

    -   sisältää paikan $\mathbf{x}$ ja

    -   ajanhetki kokonaislukuna.

    Tuloksia verrataan tunnettuun tarkkaan ratkaisuun ja etäisyys
    oikeasta saa olla kullakin askeleella enintään $0.01$. Laskenta ei
    saa kestää yli $5\,\rm{s}$.

-   **M32.** [Kitkaton
    jousi-massa-systeemi](https://en.wikipedia.org/wiki/Harmonic_oscillator#Simple_harmonic_oscillator)
    koostuu kappaleeseen (massa $m=2$) kiinnitetystä jousesta
    (jousivakio $k=3$). Usein näissä järjestelmissä jätetään painovoima
    huomioimatta, ja niin toimitaan nytkin. Koordinaatisto asetetaan
    siten, että kun kappaleen paikka $x=0$, niin kappaleeseen ei vaikuta
    voimaa. Jos kappaletta poikkeutetaan tästä paikasta, niin tällöin
    kappaleeseen vaikuttaa voima $F=-kx$.  
    Olkoon alkutilanteessa $t=0$, $x=0.5$ ja $v=0$. Kirjoita ohjelma,
    joka kirjoittaa tiedostoon `output.txt` kappaleen paikan sekunnin
    välein ajanjaksona $0\dots100$ sekuntia.

-   **M33.** Kirjoita ohjelma, joka tulostaa viestin "Olen antanut
    kurssipalautteen."
