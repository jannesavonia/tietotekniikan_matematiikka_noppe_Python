# Matriisit

## Käänteismatriisi

<hr class="theorem-block-start" style="border:0; border-top:4px solid #1565c0; margin-top:1.2em; margin-bottom:0.6em;">

#### Definition

Olkoon $\mathbf{A} \in \mathbb{R}^{n \times n}$, siis neliömatriisi. Mikäli on
olemassa matriisi $\mathbf{A}^{-1} \in \mathbb{R}^{n \times n}$ siten, että
$$\mathbf{A}\mathbf{A}^{-1} = \mathbf{I} \ ,$$ niin $\mathbf{A}^{-1}$ on tällöin $\mathbf{A}$:n
*käänteismatriisi* ja matriisin $\textbf{A}$ sanotaan olevan kääntyvä
(tai säännöllinen).

<hr class="theorem-block-end" style="border:0; border-top:2px solid #1565c0; margin-top:0.6em; margin-bottom:1.2em;">

<hr class="theorem-block-start" style="border:0; border-top:4px solid #6a1b9a; margin-top:1.2em; margin-bottom:0.6em;">

#### Theorem

Mikäli käänteismatriisi on olemassa, niin

-   $\mathbf{A}\mathbf{A}^{-1} = \textbf{I} = \mathbf{A}^{-1}\mathbf{A}$,

-   $\mathbf{A}^{-1}$ on yksikäsitteinen ja

-   ${(\mathbf{A}^{-1})}^{-1} = \mathbf{A}$.

<hr class="theorem-block-end" style="border:0; border-top:2px solid #6a1b9a; margin-top:0.6em; margin-bottom:1.2em;">

<hr class="theorem-block-start" style="border:0; border-top:4px solid #ef6c00; margin-top:1.2em; margin-bottom:0.6em;">

#### Example 47

Olkoon
$\mathbf{A}=\operatorname{diag}(a_1,\,a_2,\, \dots,\,a_n)$, $a_i\ne 0$ kaikilla $i$.
Etsitään, jos on olemassa, $\mathbf{B} \in \mathbb{R}^{n \times n}$ s.e
$\mathbf{A} \mathbf{B} = \mathbf{I}$.  
Lasketaan 

$$
\begin{aligned}
        \mathbf{I} = \mathbf{A} \mathbf{B} & = \begin{pmatrix}
            a_1 b_{1,1} & a_1 b_{1,2} \dots &a_1 b_{1,n}\\
            a_2 b_{2,1} & a_2 b_{2,2} \dots &a_2 b_{2,n}\\
            \vdots & \vdots \ddots & \vdots \\
            a_n b_{n,1} & a_n b_{n,2} \dots & a_n b_{n,n}\ ,
        \end{pmatrix}
\end{aligned}
$$

 joten on oltava $b_{i,j}=0$ kun $i\ne j$ ja
$b_{i,i}=\frac{1}{a_i}$.  
Siis
$\mathbf{A}^{-1} = diag(\frac{1}{a_1},\, \frac{1}{a_2},\, \dots,\, \frac{1}{a_n})$.

<hr class="theorem-block-end" style="border:0; border-top:2px solid #ef6c00; margin-top:0.6em; margin-bottom:1.2em;">

#### Example 48

Matriisi
$\mathbf{R}(\alpha) = \begin{pmatrix} \cos \alpha & -\sin \alpha \\ \sin \alpha & \cos \alpha\end{pmatrix} = \begin{pmatrix}c & -s \\ s & c\end{pmatrix}$
kiertää tason pisteitä kulman $\alpha$ verran.  
Etsitään, jos on olemassa, $\mathbf{B} \in \mathbb{R}^{2 \times 2}$ s.e
$\mathbf{R}(\alpha) \mathbf{B}=\mathbf{I}$.  
Lasketaan 

$$
\begin{aligned}
        \mathbf{I} = \mathbf{R}(\alpha) \mathbf{B} & = \begin{pmatrix}
            c b_{1,1} - s b_{2,1} & c b_{1,2} - s b_{2,2} \\
            s b_{1,1} + c b_{2,1} & s b_{1,2} + c b_{2,2} \\
        \end{pmatrix}\ ,
\end{aligned}
$$

joka toteutuu kun

> *Muista: $\sin^2 \alpha + \cos^2 \alpha=1$,
> $\sin -\alpha = - \sin \alpha$ ja $\cos -\alpha = \cos \alpha$*

$\mathbf{B}=\begin{pmatrix}c & s \\ -s & c\end{pmatrix}=\mathbf{R}(-\alpha)$.  
Siis $\mathbf{R}(\alpha)^{-1} = \mathbf{R}(-\alpha)$.
