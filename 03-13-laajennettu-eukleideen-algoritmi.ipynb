## Laajennettu Eukleideen algoritmi

### Laajennettu Eukleideen algoritmi

-   [Algoritmin](https://cp-algorithms.com/algebra/extended-euclid-algorithm.html)
    avulla saadaan ratkaistua $\operatorname{syt}\left(a,\,b\right)$ sekä $x, y \in \mathbb{Z}$
    siten että $ax+by=\operatorname{syt}\left(a,\,b\right)$

<hr class="theorem-block-start" style="border:0; border-top:4px solid #0277bd; margin-top:1.2em; margin-bottom:0.6em;">

#### Algorithm: Laajennettu Eukleideen algoritmi

Olkoon $a\ge b>0$

1.  Asetetaan $\hat{a}=a$ ja $\hat{b}=b$

2.  Asetetaan $x=1$, $y=0$, $x_1=0$, $y_1=1$

3.  Ratkaise $r$ ja $q$ jakoyhtälöstä $\hat{a} = q\hat{b}+r$

    -   Pythonilla <span class="nodecor">`r=a%b`</span> ja
        <span class="nodecor">`q=a//b`</span>

4.  Asetetaan $x_0=x$, $x=x_1$ ja $x_1=x_0-qx_1$

5.  Asetetaan $y_0=y$, $y=y_1$ ja $y_1=y_0-qy_1$

6.  Jos $r=0$, niin tällöin $\operatorname{syt}\left(a,\,b\right)=\hat{b}$ ja lisäksi $x$ ja $y$
    toteuttavat $ax+by=\operatorname{syt}\left(a,\,b\right)$. Palauta siis $\hat{b}$, $x$ ja
    $y$.

7.  Asetetaan $\hat{a} = \hat{b}$ ja $\hat{b} = r$

8.  Jatka kohdasta (3)
