# Differentiaaliyhtälöiden numeerinen ratkaiseminen

## Harjoituksia

### Harjoituksia

-   **M26.** Olkoon nopeus $\mathbf{v}(t)=\begin{pmatrix}-\sin t \\ \cos t\end{pmatrix}$ ja
    $\mathbf{x}(0) = \begin{pmatrix}1 \\ 0\end{pmatrix}$. Kirjoita funktio
    `euler_circle(Dt, n)`, joka kirjoittaa tekstitiedostoon `output.txt`
    jokaisella aika-askelella:

    -   $t_i$ = ajanhetki, ensimmäinen `0` ja viimeinen `n Dt`,

    -   $x_i$ = ajanhetkeä vastaava $x$-koordinaatti sekä

    -   $y_i$ = ajanhetkeä vastaava $y$-koordinaatti.

    Sarakkeet erotetaan välilyönnillä. Tiedostossa on siis `n+1` riviä.
    Parametrit:

    -   `Dt` = aika-askel, $\Delta t$ ja

    -   `n` = aika-askeleiden lukumäärä

    Ratkaisussa käytetään luentokalvoissa esitettyä Euler-menetelmää.
    Tarkempi määrittely selviää koodipohjasta.

    \end{aligned}
$$

 Kirjoita funktio `euler_spiral(Dt, n)`, joka
