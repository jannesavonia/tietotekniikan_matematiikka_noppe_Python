# Vektorit

## Kohtisuoruus ja vektoreiden välinen kulma

<hr class="theorem-block-start" style="border:0; border-top:4px solid #6a1b9a; margin-top:1.2em; margin-bottom:0.6em;">

#### Theorem

Olkoon $\mathbf{x}$ ja $\mathbf{y} \in \mathbb{R}^n$ ja
$\mathbf{x} \neq \mathbf{0} \neq \mathbf{y}$. Jos $\mathbf{x} \cdot \mathbf{y} = 0$, niin
tällöin $\cos \gamma = 0$.

<hr class="theorem-block-end" style="border:0; border-top:2px solid #6a1b9a; margin-top:0.6em; margin-bottom:1.2em;">

<hr class="theorem-block-start" style="border:0; border-top:4px solid #ad1457; margin-top:1.2em; margin-bottom:0.6em;">

#### Corollary

Tällöin $\gamma = \pm \frac{1}{2} \pi$, eli vektorit ovat toisiaan
vastaan kohtisuorassa.

#### Theorem

Olkoon $\mathbf{x}$ ja $\mathbf{y} \in \mathbb{R}^n$ ja
$\mathbf{x} \neq \mathbf{0} \neq \mathbf{y}$. Tällöin
$$\cos \gamma = \frac{\mathbf{x}\cdot\mathbf{y}}{\lVert \mathbf{x} \rVert\lVert \mathbf{y} \rVert}\ .$$

<hr class="theorem-block-end" style="border:0; border-top:2px solid #6a1b9a; margin-top:0.6em; margin-bottom:1.2em;">

<hr class="theorem-block-start" style="border:0; border-top:4px solid #00695c; margin-top:1.2em; margin-bottom:0.6em;">

#### Remark

Edellisestä saadaan vektoreiden välinen kulma ottamalla puolittain
$\arccos$.

<hr class="theorem-block-end" style="border:0; border-top:2px solid #00695c; margin-top:0.6em; margin-bottom:1.2em;">

<hr class="theorem-block-start" style="border:0; border-top:4px solid #ad1457; margin-top:1.2em; margin-bottom:0.6em;">

#### Corollary

Yksikkövektoreille $\hat{\mathbf{x}},\hat{\mathbf{y}} \in \mathbb{R}^n$
$$\cos \gamma = \hat{\mathbf{x}}\cdot\hat{\mathbf{y}}\ .$$

<hr class="theorem-block-end" style="border:0; border-top:2px solid #ad1457; margin-top:0.6em; margin-bottom:1.2em;">

<hr class="theorem-block-start" style="border:0; border-top:4px solid #ef6c00; margin-top:1.2em; margin-bottom:0.6em;">

#### Example 31

Olkoon $\mathbf{x}=\begin{pmatrix}1&2&-1&3\end{pmatrix}$ ja $\mathbf{y}=\begin{pmatrix}2&-2&-1&1\end{pmatrix}$.
Tällöin 

$$
\begin{aligned}
            \mathbf{x} \cdot \mathbf{y} & = 1\cdot2 -2\cdot 2 + 1\cdot 1 + 3 \cdot 1 = 2\\
            \lVert \mathbf{x} \rVert & = \sqrt{1^2+2^2+1^2+3^2} = \sqrt{15}\\
            \lVert \mathbf{y} \rVert & = \sqrt{2^2+2^2+1^2+1^2} = \sqrt{10}\\
            \cos \gamma & = \frac{\mathbf{x} \cdot \mathbf{y}}{\lVert \mathbf{x} \rVert\lVert \mathbf{y} \rVert}\\
            & = \frac{2}{\sqrt{150}}\\
            \Rightarrow \gamma & = \arccos \frac{2}{\sqrt{150}} \approx 1.407\ .
\end{aligned}
$$

<hr class="theorem-block-end" style="border:0; border-top:2px solid #ef6c00; margin-top:0.6em; margin-bottom:1.2em;">

<hr class="theorem-block-start" style="border:0; border-top:4px solid #ef6c00; margin-top:1.2em; margin-bottom:0.6em;">

#### Example 32

Olkoon $\mathbf{x}=\begin{pmatrix}1&2&-3&-1\end{pmatrix}$ ja $\mathbf{y}=\begin{pmatrix}2&-1&-1&3\end{pmatrix}$.
Tällöin 

$$
\begin{aligned}
            \mathbf{x} \cdot \mathbf{y} & = 1\cdot2 +2\cdot (-1) + (-3)\cdot (-1) + (-1) \cdot 3 = 0\\
            \Rightarrow \cos \gamma & = 0 \\
            \Rightarrow \gamma & = \pm \frac{\pi}{2}\ .
\end{aligned}
$$
