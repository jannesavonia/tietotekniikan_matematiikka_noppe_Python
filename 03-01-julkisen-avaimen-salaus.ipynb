### Julkisen avaimen salaus

-   Idea:

    -   Vastaanottaja generoi avainparin $K_{pub}$ ja $K_{priv}$

    -   Viesti salataan julkisella avaimella $K_{pub}$

    -   Viestiä ei voi purkaa julkisella avaimella

    -   Viestin purkamiseen vaaditaan salainen avain $K_{priv}$

-   Vastaanottaja lähettä julkisen avaimen lähettäjälle

-   Jos $K_{pub}$ vuotaa ulkopuolisen käsiin, ei siitä ole haittaa

-   Vastaanottaja säilyttää itse salaisen avaimen $K_{priv}$ ja purkaa
    salatun viestin sillä
