# Matriisit

## Matriisityyppejä

<hr class="theorem-block-start" style="border:0; border-top:4px solid #1565c0; margin-top:1.2em; margin-bottom:0.6em;">

#### Definition

Olkoon $\mathbf{A} \in \mathbb{R}^{m \times n}$.

-   Jos $m=n$, niin $\mathbf{A}$:n sanotaan olevan **neliömatriisi**.

    -   Neliömatriisit $\mathbb{R}^{3 \times 3}$ ja $\mathbb{R}^{4 \times 4}$
        ovat keskeisessä asemassa tietokonegrafiikassa.

-   Jos $n=1$ sanotaan matriisia **pystyvektoriksi**.

    -   Samaistetaan jatkossa $\mathbb{R}^m$:n vektorit ja
        $\mathbb{R}^{m \times 1}$ matriisit.

-   Jos $m=1$ sanotaan matriisia **vaakavektoriksi**.

-   Jos $n=m=1$, niin samaistetaan $\mathbf{A}$ skalaarin
    $a_{1,1} \in \mathbb{R}$ kanssa.

<hr class="theorem-block-end" style="border:0; border-top:2px solid #1565c0; margin-top:0.6em; margin-bottom:1.2em;">

<hr class="theorem-block-start" style="border:0; border-top:4px solid #ef6c00; margin-top:1.2em; margin-bottom:0.6em;">

#### Example 40

-   Neliömatriisi: 

$$
\begin{pmatrix}
                    1 & 2 \\
                    3 & 4
                \end{pmatrix}
$$

-   Vaakavektori: 

$$
\begin{pmatrix}
                    8 & 2 & 0 & 1\\
                \end{pmatrix}
$$

-   Pystyvektori: 

$$
\begin{pmatrix}
                    8 \\ 2 \\ 0 \\1
                \end{pmatrix}
$$

-   Skalaari: 

$$
\begin{pmatrix}
                    4.3
                \end{pmatrix} = 4.3
$$

#### Definition

Matriisin $\mathbf{A}\in \mathbb{R}^{m \times n}$ alkioita $a_{ij}$, joissa
$i=j$ kutsutaan $\mathbf{A}$:n **diagonaalialkioiksi**. Mikäli matriisin
kaikki alkiot lukuunottamatta diagonaalialkioita ovat $0$, sanotaan
matriisia **diagonaalimatriisiksi**.

<hr class="theorem-block-end" style="border:0; border-top:2px solid #1565c0; margin-top:0.6em; margin-bottom:1.2em;">

<hr class="theorem-block-start" style="border:0; border-top:4px solid #ef6c00; margin-top:1.2em; margin-bottom:0.6em;">

#### Example 41

$\begin{pmatrix}
        1 & 0 & 0 \\
        0 & 2 & 0 \\
        0 & 0 & 1 
    \end{pmatrix}$ , $\begin{pmatrix}
        1 & 0 & 0 \\
        0 & 2 & 0 
    \end{pmatrix}$ ja $\begin{pmatrix}
        1 & 0 & 0 \\
        0 & 2 & 0 
    \end{pmatrix}^T$ ovat diagonaalimatriiseja.

<hr class="theorem-block-end" style="border:0; border-top:2px solid #ef6c00; margin-top:0.6em; margin-bottom:1.2em;">

<hr class="theorem-block-start" style="border:0; border-top:4px solid #455a64; margin-top:1.2em; margin-bottom:0.6em;">

#### Notation

Diagonaalimatriisia, joka on myös neliömatriisi, voidaan merkitä
lyhyemmin 

$$
\begin{pmatrix}
            a_{1,1} & 0 & \dots & 0 \\
            0 & a_{2,2} & \dots & 0 \\
            \vdots & \vdots & \ddots & \vdots \\
            0 & 0 & \dots & a_{n,n}
        \end{pmatrix}
        =
        \begin{pmatrix}
            a_1 & 0 & \dots & 0 \\
            0 & a_2 & \dots & 0 \\
            \vdots & \vdots & \ddots & \vdots \\
            0 & 0 & \dots & a_n
        \end{pmatrix}
        =
        \operatorname{diag}(a_1, a_2, \dots, a_n).
$$

<hr class="theorem-block-end" style="border:0; border-top:2px solid #455a64; margin-top:0.6em; margin-bottom:1.2em;">

<hr class="theorem-block-start" style="border:0; border-top:4px solid #ef6c00; margin-top:1.2em; margin-bottom:0.6em;">

#### Example 42

$$
\begin{aligned}
        \operatorname{diag}(3,\,-\frac{1}{2},\, \pi) & = \begin{pmatrix}
            3 & 0 0\\
            0 & -\frac{1}{2} 0 \\
            0 & 0 \pi
        \end{pmatrix}\\[1.5ex]
        \operatorname{diag}(-e^\pi, \sin(\frac{7}{5})) & = \begin{pmatrix}
            -e^\pi & 0 \\
            0 & \sin(\frac{7}{5})
        \end{pmatrix}
\end{aligned}
$$

#### Definition

Nollamatriisi 

$$
\mathbf{0}=
        \begin{pmatrix}
            0 & 0 & \dots & 0 \\
            0 & 0 & \dots & 0 \\
            \vdots & \vdots & \ddots & \vdots \\
            0 & 0 & \dots & 0
        \end{pmatrix}
        \ .
$$

<hr class="theorem-block-end" style="border:0; border-top:2px solid #1565c0; margin-top:0.6em; margin-bottom:1.2em;">

<hr class="theorem-block-start" style="border:0; border-top:4px solid #00695c; margin-top:1.2em; margin-bottom:0.6em;">

#### Remark

-   $\mathbf{0}$ on matriisiyhteenlaskun neutraalialkio, eli pätee
    $\mathbf{0}+\mathbf{A}=\mathbf{A}$.

-   Merkintää $\mathbf{0}$ käytetään sekä matriisille, että vektorille.
    Mikäli sekaannuksen vaaraa on, käytetään nollavektorille jatkossa
    merkintää $\bar{\mathbf{0}}$.

-   Nollamatriisin ei tarvitse olla neliömatriisi.
