## Homogeeninen koordinaatisto

### Yhdistetyt kuvaukset

-   Simulaatioissa kappaleet muodostuvat usein kolmioista

-   Kärkipisteiden paikkaa muokataan tarpeen mukaan.

-   Lue [Explaining basic 3D
    theory](https://developer.mozilla.org/en-US/docs/Games/Techniques/3D_on_the_web/Basic_theory)

-   Katso [Youtube - 2D
    Transformatinos](https://youtu.be/SLWCc-mcd-I?si=tCg7eehztwMJhoLp)
    tai lue [Tutorialspoint - 2D
    Transformation](https://www.tutorialspoint.com/computer_graphics/2d_transformation.htm)

-   Kiertoa, skaalausta ja peilausta vastaavat matriisit
    (=lineaarikuvaukset) tunnetaan.

-   Yhdistettyjä kuvauksia vastaa matriisien tulo

    -   Piste $\mathbf{x}$ peilataan origon suhteen ja sitten kierretään
        $\pi/2$ radiaania origon suhteen

    -   $\mathbf{y}=\mathbf{R}(\mathbf{M}(\mathbf{x})) = \mathbf{R} \mathbf{M}\mathbf{x} = \mathbf{A}\mathbf{x}$

    -   Jos muunnettavia pisteitä on useampi, kannattaa laskea ensin
        $\mathbf{A}=\mathbf{R}\mathbf{M}$

### Siirtoa vastaava matriisi

-   Halutaan muodostaa linearikuvaus $\mathbf{A}$, joka siirtää tason
    pisteitä siten, että $x_1 \mapsto x_1+1$ ja $x_2 \mapsto x_2-1$

-   Lineaarisuuden nojalla origo kuvautuu origoksi

-   $\Rightarrow$ Vaadittua lineaarikuvausta ei ole!

-   $\Rightarrow$ Ei siirtoa vastaavaa matriisia!

-   Siirto on oleellinen operaatio esimerkiksi simuloinneissa ja
    peleissä

-   Ratkaisu:

    -   Otetaan käyttöön yksi ylimääräinen dimensio

        > *Teknisesti upotetaan koordinaatisto korkeampiulotteiseen
        > koordinaatistoon*

    -   $\Rightarrow$ Voidaan välttää origon käsittely  

    -   **Homogeeninen koordinaatisto**

        > *Lisätylle $1$:lle on mahdollista määritellä muitakin
        > merkityksiä, mutta tällä kurssilla näin ei toimita*

        , jossa $\begin{pmatrix}x_1\\x_2\end{pmatrix} \sim \begin{pmatrix}x_1\\x_2\\1\end{pmatrix}$

#### Example 56

-   $\mathbb{R}$:lla ainoa lineaarikuvaus on vakiolla kertominen

-   Haluttaisiin kuvata $x \mapsto x+1$ lineaarikuvauksen avulla

-   Valitaan linearikuvaus $c$ siten, että $1 \mapsto 2$, eli
    $c\times 1 = 2 \ \Rightarrow c=2$

-   Toisaalta pitää $2 \mapsto 3$, eli
    $c \times 2 = 3 \ \Rightarrow  c=\frac{3}{2} \neq 2$

-   Tämä tiedettiin, ei voinut toimia

#### Example 56

-   Upotetaan $\mathbb{R}^1$ korkeampiulotteiseen $\mathbb{R}^2$:een

    > *Katso kuva seuraavasta kalvosta*

    siten, että $$x \sim \mathbf{x} = \begin{pmatrix}x \\ 1\end{pmatrix}$$

-   Muodostetaan lineaarikuvaus $\mathbf{T}_{\Delta x}$, jolla
    $\begin{pmatrix}x \\ 1\end{pmatrix} \mapsto \begin{pmatrix}x+\Delta x \\ 1\end{pmatrix}$

-   Valitaan $$\mathbf{T}_{\Delta x} = \begin{pmatrix}1 & \Delta x \\ 0 & 1\end{pmatrix}$$

-   Tällöin
    $$\mathbf{T}_{\Delta x} \begin{pmatrix}x \\ 1\end{pmatrix} = \begin{pmatrix}1 & \Delta x \\ 0 & 1\end{pmatrix} \begin{pmatrix}x \\ 1\end{pmatrix}= \begin{pmatrix}x + \Delta x \\ 1\end{pmatrix}$$

#### Example 56

-   Muodostetaan myös skaalaus

    > *Muita muunnoksia ei $\mathbb{R}^1$:ssä tarvitakaan.*

-   Olkoon piste $x \in \mathbb{R}$

-   Homogeenisissa koordinateissa sitä vastaa $\mathbf{x}=\begin{pmatrix}x \\ 1\end{pmatrix}$

-   Halutaan muodostaa lineaarikuvaus $\mathbf{S}_c$, jolla $x \mapsto cx$

-   Valitaan $\mathbf{S}_c=\begin{pmatrix}c & 0\\ 0 & 1\end{pmatrix}$, jolloin
    $$\mathbf{S}_c \mathbf{x} = \begin{pmatrix}c & 0\\ 0 & 1\end{pmatrix} \begin{pmatrix}x \\ 1\end{pmatrix}  = \begin{pmatrix}cx+0\cdot 1 \\ 0 \cdot x + 1 \cdot 1\end{pmatrix}  =  \begin{pmatrix}cx \\ 1\end{pmatrix}$$

#### Example 56

-   Tarkastellaan $\mathbb{R}$:n pisteitä $-2$, $0$ ja $3$

-   Näitä vastaavat homogeeniset koordinaartit ovat $\begin{pmatrix}-2\\1\end{pmatrix}$,
    $\begin{pmatrix}0\\1\end{pmatrix}$ ja $\begin{pmatrix}3\\1\end{pmatrix}$

-   Kirjoitetaan "vektorit vierekkäin" matriisiksi
    $\mathbf{X}=\begin{pmatrix}-2 & 0 & 3 \\ 1 & 1 & 1\end{pmatrix}$

-   Muodostetaan muunnos, joka

    1.  skaalaa pisteitä $2$:lla, merkitään $\mathbf{S}_2$,

    2.  siirtää pisteitä "oikealle" $1$ askeleen, merkitään $\mathbf{T_1}$
        ja

    3.  skaalaa pisteitä $\frac{1}{2}$:lla, merkitään $\mathbf{S}_{1/2}$

-   Tulosvektorit (matriisi)
    $$\mathbf{Z} = \underbrace{\mathbf{S}_{1/2}\mathbf{T_1}\mathbf{S}_2}_{=\mathbf{A}}\mathbf{X}$$

#### Example 56

-   Lasketaan ensin
    $$\mathbf{A}= \begin{pmatrix}\frac{1}{2} & 0 \\ 0 & 1\end{pmatrix} \begin{pmatrix}1 & 1 \\ 0 & 1\end{pmatrix} \begin{pmatrix}2 & 0 \\ 0 & 1\end{pmatrix}  = \begin{pmatrix}\frac{1}{2} & 0 \\ 0 & 1\end{pmatrix} \begin{pmatrix}2 & 1 \\ 0 & 1\end{pmatrix}  = \begin{pmatrix}1 & \frac{1}{2} \\ 0 & 1\end{pmatrix}$$

-   Muunnetaan pisteet
    $$\mathbf{Z} = \mathbf{A} \mathbf{X} = \begin{pmatrix}1 & \frac{1}{2} \\ 0 & 1\end{pmatrix} \begin{pmatrix}-2 & 0 & 3 \\ 1 & 1 & 1\end{pmatrix}  = \begin{pmatrix} -\frac{3}{2} & \frac{1}{2} & \frac{7}{2} \\ 1 & 1 &1\end{pmatrix}$$

-   Muunnetut pisteet ovat siis $-\frac{3}{2}$, $\frac{1}{2}$ ja
    $\frac{7}{2}$

#### Remark

Sama periaate toimii myös $\mathbb{R}^2$:ssa ja $\mathbb{R}^3$:ssa, vaikka
mahdollisten muunnosten määrä kasvaakin kiertojen myötä.
