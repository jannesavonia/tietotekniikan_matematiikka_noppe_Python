# Kurssin sisältö

### Yleistä

-   [Kurssikuvaus](https://opinto-opas.peppi.savonia.fi/realization/12952)

-   Laajuus: 5 op

-   Osaamistavoitteet – OPS

-   Toteutustavat:

    -   (Lähiopetus)

    -   Videot

    -   Omatoimista opiskelua

    -   Ohjelmointiharjoituksia Python- tai C#-kielellä

    -   Harjoitustehtäviä

-   Keskeiset sisällöt:

    -   Lukuteoria ja RSA-salaus

    -   Lineaarialgebra

    -   Differentiaaliyhtälöiden numeerinen ratkaiseminen

-   Suoritustavat:

    -   Palautettavat ohjelmointitehtävät, N kpl

-   Kurssi on avoinna 31.7.2027 saakka

-   Arviointi:

    -   $30 \mapsto 5$

    -   $26 \mapsto 4$

    -   $23 \mapsto 3$

    -   $20 \mapsto 2$

    -   $16 \mapsto 1$
