# Lukuteoriaa

## Harjoituksia

### Harjoituksia

-   **10.** Laske $13-17 \pmod{4}$

-   **11.** Laske $13 \cdot 17 \pmod{4}$

-   **12.** Laske $13^3 \cdot 17^3 \pmod{4}$

-   **M9.** Toteuta modulaariaritmetiikka-luokka

    -   `*`, `+` ja `-`

    -   `a**k`, kun $k\in \mathbb{N}$

        -   Jos $k$ ei ole luonnollinen luku, niin heitetään poikkeus

    -   `==` toimii kuten kongruenssi-operaattori

    -   $n$ annetaan muodostimen yhteydessä

    -   Jos operaattorin eri puolilla on eri $n$, niin heitetään
        poikkeus

Vastaukset laskuharjoituksiin ovat (ei välttämättä samassa
järjestyksessä) $2 \pmod{4}$, $1 \pmod 4$ ja $0 \pmod{4}$
