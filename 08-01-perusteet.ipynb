# Differentiaaliyhtälöiden numeerinen ratkaiseminen

## Perusteet

### Taustaa

-   Differentiaaliyhtälöiden ratkaiseminen käsin on varsin työlästä ja
    virhealtista.

-   Usein 1- ja 2-kl lineaarinen DY voidaan ratkaista analyyttisesti
    myös koneella.

-   Epälineaaristen DY:den ratkaiseminen on erittäin hankalaa ja
    kattavaa teoriaa niiden ratkaisemiseksi analyyttisesti ei ole

    > *Ainakaan minun tietojeni mukaan*

    .

-   Joskus yhtälön oikea puoli voi riippua esimerkiksi käyttäjän
    antamasta syötteestä, tai olla muuten hankala, jolloin yhtälöä ei
    voida ratkaista etukäteen.

-   $\Rightarrow$ Numeerinen ratkaiseminen

#### Remark

Jatkossa esitellään menetelmiä, joiden avulla voidaan ratkaista
numerisesti differentiaaliyhtälöitä, jotka ovat muotoa

$$
\tag{1}
            \mathbf{y}'(t) = \mathbf{f}(t, \mathbf{y}(t))\ ,
$$

 alkuehdolla
$\mathbf{y}(t_0)=\mathbf{y}_0$ .

> *Käytetään tässä muuttujana $t$:tä, koska näin usein on. Tilanne ei
> muutu miksikään, vaikka käytetäisiin $x$:ää.*
