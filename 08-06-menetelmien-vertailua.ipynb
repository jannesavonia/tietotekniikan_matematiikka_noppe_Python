# Differentiaaliyhtälöiden numeerinen ratkaiseminen

## Menetelmien vertailua

### 2-kl lineaarinen yhtälö

-   Olkoon $y\textnormal{''}(t) + y\textnormal{'}(t)+ y(t) = 0$
    alkuehdolla $y(0)=1$ ja $y\textnormal{'}(0)=0$

-   Ratkaisu

    > *[MathDF](https://mathdf.com/dif/)*

    on
    $y(t)= e^{-\frac{1}{2}t} (\cos (\frac{\sqrt{3}t}2) + \frac{1}{\sqrt{3}}\sin (\frac{\sqrt{3}t}2))$

-   Numeerisiin ratkaisuihin lisätään apumuuttuja
    $v(t)=y\textnormal{'}(t)$

-   Sijoitetaan tämä alkuperäiseen yhtälöön: 

$$
\begin{aligned}
                0 & = y\textnormal{''}(t) + y\textnormal{'}(t)+ y(t) \\
                & = v\textnormal{'}(t) + v(t)+ y(t) \\
                \Rightarrow v\textnormal{'}(t) & = - v(t) - y(t)
    \end{aligned}
$$

-   Tällöin yhtälö saa muodon
    $$\begin{pmatrix}y\textnormal{'}(t) \\ v\textnormal{'}(t)\end{pmatrix} = \begin{pmatrix}v(t) \\ -v(t) - y(t)\end{pmatrix}$$

<hr class="code-block-start" style="border:0; border-top:4px solid #455a64; margin-top:1.2em; margin-bottom:0.6em;">

```python
%reset -f

# Source file: examples/compare01.py

import numpy as np 
from matplotlib import pyplot as plt 

def y(t):
    return np.exp(-0.5*t)*(np.cos((np.sqrt(3)/2)*t)+(1/np.sqrt(3))*np.sin((np.sqrt(3)/2)*t))
```

```python
def f(t, z):
    y=z[0]
    v=z[1]
    return np.array([v, -v-y])
```

```python
N_steps=50
t_values=np.linspace(0.0, 10.0, num=N_steps+1)
Dt=(t_values[1]-t_values[0])
y_exact=y(t_values)

z=np.array([1, 0])
y_euler=[z[0]]
for prev_t in t_values[0:-1]:
    z=z+Dt*f(prev_t, z)
    y_euler.append(z[0])
```

```python
z=np.array([1, 0])
y_ab=[z[0]]
f_list=[f(0, z)]
```

```python
for prev_t in t_values[0:-1]:
    f_list.append(f(prev_t, z))
    if len(f_list)<3:
        z=z+Dt*f_list[-1]
    else:
        z=z+(Dt/12)*(23*f_list[-1]-16*f_list[-2]+5*f_list[-3])
        f_list.pop(0)
    y_ab.append(z[0])
```

```python
plt.plot(t_values, y_euler, label='Euler')
plt.plot(t_values, y_ab, label='AB')
plt.plot(t_values, y_exact, 'k:', label='exact')
plt.legend()
plt.show()

plt.title('Error')
plt.plot(t_values, np.abs(y_euler-y_exact), label='Euler')
plt.plot(t_values, np.abs(y_ab-y_exact), label='AB')
plt.yscale('log')
plt.legend()
plt.show()
```

### Ei-vakiokertoiminen DY

-   Olkoon $y(0)=1$ ja $y\textnormal{'}(t)=y(t)\sin^2 t$.

-   Jaetaan $y$:llä ja integroidaan puolittain: 

$$
\begin{aligned}
                \int \frac{y\textnormal{'}(t)}{y(t)}\, dt & = \int \sin^2 t\, dt\\
                \Rightarrow\ln \lvert y(t) \rvert + C & = \int \frac{1}{2} - \frac{1}{2} \cos 2t\, dt
                =  \frac{1}{2} (t- \frac{1}{2} \sin 2t) \\
                \Rightarrow \lvert y(t) \rvert &= \exp(\frac{1}{2} (t- \frac{1}{2} \sin 2t)-C) \\  \Rightarrow y(t) &=& \exp(\frac{1}{2} (t- \frac{1}{2} \sin 2t)-C) \\
                \Rightarrow 1 &= y(0)  =  \exp(\frac{1}{2} (0- \frac{1}{2} \sin 2\cdot0)-C) \\
                \Rightarrow 0 & = \frac{1}{2} (0- \frac{1}{2} \sin 0)-C \ \
                \Rightarrow C = 0\ .
    \end{aligned}
$$

-   Tarkka ratkaisu on siis
    $y(t) = \exp (\frac{1}{2}t- \frac{1}{4} \sin 2t)$ .

```python
%reset -f

# Source file: examples/compare02.py

import numpy as np 
from matplotlib import pyplot as plt 

def y(t):
    return np.exp(0.5*(t-0.5*np.sin(2*t)))
```

```python
def f(t, y):
    return y*np.sin(t)*np.sin(t)
```

```python
N_steps=100
t_values=np.linspace(0.0, 30.0, num=N_steps+1)
Dt=(t_values[1]-t_values[0])
y_exact=y(t_values)
```

```python
y=1
y_euler=[y]
for t_prev in t_values[0:-1]:
    y=y+Dt*f(t_prev, y)
    y_euler.append(y)
```

```python
y=1
y_ab=[y]
f_list=[f(0, y)]
for t_prev in t_values[0:-1]:
    f_list.append(f(t_prev, y))
    if len(f_list)<3:
        y=y+Dt*f_list[-1]
    else:
        y=y+(Dt/12)*(23*f_list[-1]-16*f_list[-2]+5*f_list[-3])
        f_list.pop(0)
    y_ab.append(y)
```

```python
plt.plot(t_values, y_euler, label='Euler')
plt.plot(t_values, y_ab, label='AB')
plt.plot(t_values, y_exact, 'k:', label='exact')
plt.legend()
plt.show()

plt.title('Error')
plt.plot(t_values, np.abs(y_euler-y_exact), label='Euler')
plt.plot(t_values, np.abs(y_ab-y_exact), label='AB')
plt.yscale('log')
plt.legend()
plt.show()
```
