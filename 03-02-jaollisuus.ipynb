# Lukuteoriaa

## Jaollisuus

#### Remark

Tässä luvussa on paljon teoreettista asiaa, jonka syvällinen
ymmärtäminen ei ole aivan välttämätöntä salauksen toiminnan
ymmärtämiseksi. Asiat on kuitenkin esitelty, jotta asiasta
syvällisemmin kiinnostuneet voivat tutustua niihin.

<hr class="theorem-block-start" style="border:0; border-top:4px solid #1565c0; margin-top:1.2em; margin-bottom:0.6em;">

#### Definition

-   **Luonnolliset luvut** $$\mathbb{N}=\{1,\, 2,\, 3\, \dots\}$$

-   **Kokonaisluvut**
    $$\mathbb{Z}=\{\dots,\,-2,\,-1,\,0,\, 1,\, 2,\dots\}$$

<hr class="theorem-block-end" style="border:0; border-top:2px solid #1565c0; margin-top:0.6em; margin-bottom:1.2em;">

<hr class="theorem-block-start" style="border:0; border-top:4px solid #6a1b9a; margin-top:1.2em; margin-bottom:0.6em;">

#### Theorem

Jokaiselle $a,\, b \in \mathbb{Z}$, kun $b>0$ löytyy yksikäsitteiset
$q,\,r \in \mathbb{Z}$ siten, että
$$a=qb+r\ \textnormal{ ja }\ 0 \le r <b\ .$$

<hr class="theorem-block-end" style="border:0; border-top:2px solid #6a1b9a; margin-top:0.6em; margin-bottom:1.2em;">

<hr class="theorem-block-start" style="border:0; border-top:4px solid #00695c; margin-top:1.2em; margin-bottom:0.6em;">

#### Remark

Tutummin tämä lause sanoo, että kun operoidaan kokonaisluvuilla, niin
osamäärän $a/b$ kokonaislukuosa $q$ ja jakojäännös $r$ ovat olemassa
ja yksikäsitteiset.

<hr class="theorem-block-end" style="border:0; border-top:2px solid #00695c; margin-top:0.6em; margin-bottom:1.2em;">

<hr class="theorem-block-start" style="border:0; border-top:4px solid #ef6c00; margin-top:1.2em; margin-bottom:0.6em;">

#### Example 3

-   "$3/2$":

    -   $a=3$, $b=2$

    -   $3=1\cdot2+1$

    -   $q=1$ ja $r=1$

-   "$77/16$":

    -   $a=77$, $b=16$

    -   $77=4\cdot16+13$

    -   $q=4$ ja $r=13$

-   "$(-3)/2$":

    -   $a=-3$, $b=2$

    -   $-3=-2\cdot2+1$

    -   $q=-1$ ja $r=1$

<hr class="theorem-block-end" style="border:0; border-top:2px solid #ef6c00; margin-top:0.6em; margin-bottom:1.2em;">

<hr class="theorem-block-start" style="border:0; border-top:4px solid #1565c0; margin-top:1.2em; margin-bottom:0.6em;">

#### Definition

Jos kokonaisluvuille $a$ ja $b$ on olemassa $q \in \mathbb{Z}$ siten,
että $$a = qb\ ,$$ niin tällöin $b$ **jakaa** $a$:n, tai
vaihtoehtoisesti $a$ on **jaollinen** $b$:llä. Tätä merkitään
$b \vert a$.  
Jos $a$ ei ole jaollinen $b$:llä, merkitään $b \nmid a$.

<hr class="theorem-block-end" style="border:0; border-top:2px solid #1565c0; margin-top:0.6em; margin-bottom:1.2em;">

<hr class="theorem-block-start" style="border:0; border-top:4px solid #00695c; margin-top:1.2em; margin-bottom:0.6em;">

#### Remark

Edellinen sanoo siis, että jos $a/b$ jakojäännös $r=0$, niin tällöin
$b \vert a$.

<hr class="theorem-block-end" style="border:0; border-top:2px solid #00695c; margin-top:0.6em; margin-bottom:1.2em;">

<hr class="theorem-block-start" style="border:0; border-top:4px solid #1565c0; margin-top:1.2em; margin-bottom:0.6em;">

#### Definition

Kokonaisluku $b$ on kokonaisluvun $a$ **tekijä**, jos $b \vert a$.

<hr class="theorem-block-end" style="border:0; border-top:2px solid #1565c0; margin-top:0.6em; margin-bottom:1.2em;">

<hr class="theorem-block-start" style="border:0; border-top:4px solid #ef6c00; margin-top:1.2em; margin-bottom:0.6em;">

#### Example 4

-   $1 \vert 45$, koska $45=45\cdot 1$

-   $9 \vert 45$, koska $45=5\cdot 9$

-   $5 \vert 45$, koska $45=9 \cdot 5$

-   $16 \nmid 45$, koska $45 = 2 \cdot 16 + 13$, eli jakojäännös
    $b=13$

-   $78364 \vert 0$, koska $0 = 0 \cdot 78364$

-   $0 \nmid 11$, koska $11 \neq 0 = z\cdot 0$ kaikille
    $z \in \mathbb{Z}$

-   $5 \vert -10$, koska $-10 = -2 \cdot 5$

-   $2$ on luvun $16$ tekijä, koska $2 \vert 16$

-   $9$ on luvun $-81$ tekijä, koska $9 \vert -81$
