## Eulerin menetelmä

#### Example 57

Olkoon $$3 y(t)+\sin t -y\textnormal{'}(t)=0$$ alkuehdolla $y(0)=1$.  
Tämän ratkaisu on
$$y(t)=\frac{11}{10}e^{3t}- \frac{3}{10} \sin t - \frac{1}{10} \cos t\ .$$

<hr class="theorem-block-end" style="border:0; border-top:2px solid #ef6c00; margin-top:0.6em; margin-bottom:1.2em;">

### Johtaminen

Aloitetaan derivaatan määritelmästä
$$\mathbf{y}'(t) = \lim_{\Delta t \rightarrow 0} \frac{\mathbf{y}(t+\Delta t)-\mathbf{y}(t)}{\Delta t}\ .$$

Oletetaan, että $\Delta t$ on pieni, mutta $\Delta t > 0$. Tällöin
$$\mathbf{y}'(t) \approx \frac{\mathbf{y}(t+\Delta t)-\mathbf{y}(t)}{\Delta t}\ .$$

Toisaalta yhtälön (1)
merkinnöillä
$$\mathbf{f}(t, \mathbf{y}(t)) = \mathbf{y}'(t) \approx \frac{\mathbf{y}(t+\Delta t)-\mathbf{y}(t)}{\Delta t}\ ,$$
josta voidaan ratkaista
$$\mathbf{y}(t+\Delta t) \approx \mathbf{y}(t) + \Delta t\, \mathbf{f}(t, \mathbf{y}(t))\ .$$

<img src="notebook_images/tikz-8086e489409b.svg" style="width:845px; max-width:100%; height:auto;" />

Kiinnitetään $\Delta t$ ja $t_0$. Jaetaan $t$ lyhyisiin $\Delta t$:n
mittaisiin osiin, jolloin
$$t \in \{\underbrace{t_0+0\cdot\Delta t}_{=t_0},\ \underbrace{t_0+1\cdot\Delta t}_{=t_1},\ \underbrace{t_0+2\cdot\Delta t}_{t_2},\ \dots\, \underbrace{t_0+n\cdot\Delta t}_{=t_n},\ \dots \}$$
Merkitään edelliseen

> *Kiinnitetään $t=t_n$.*

-   $\mathbf{y}(t) = \mathbf{y}(t_n) = \mathbf{y}_n$,

-   $\mathbf{y}(t+\Delta t) = \mathbf{y}(t_n+\Delta t) = \mathbf{y}_{n+1}$ ja

-   $\mathbf{f}(t, \mathbf{y}(t))= \mathbf{f}(t_n, \mathbf{y}_n)=\mathbf{f}_n$ .

Tällöin saadaan rekursiokaava 

$$
\begin{aligned}
        \mathbf{y}_{n+1} & \approx \mathbf{y}_n + \Delta t \mathbf{f}_n \ \textnormal{ kun} \\
    \mathbf{y}_0 & = \mathbf{y}(0)\ .
    \end{aligned}
$$

#### Algorithm: Eulerin menetelmä

Olkoon DY yhtälön (1)
muotoa. Tällöin kiinnittämällä $\mathbf{y}_0$ ja valitsemalla askelpituus
$\Delta t$ saadaan DY:n ratkaisuja approksimoitua pisteissä
$t_0+n\Delta n$, kun $n \in \mathbb{N}$, kaavalla
$$\mathbf{y}_{n+1} = \mathbf{y}_n + \Delta t \mathbf{f}_n \ .$$ Tästä saadaan
$\mathbf{y}(t_0+n\Delta t) \approx \mathbf{y}_{n}$.

<hr class="theorem-block-end" style="border:0; border-top:2px solid #0277bd; margin-top:0.6em; margin-bottom:1.2em;">

<hr class="theorem-block-start" style="border:0; border-top:4px solid #00695c; margin-top:1.2em; margin-bottom:0.6em;">

#### Remark

Menetelmällä ei siis saada laskettua approksimaatiota kaikilla
$t\ge t_0$, vaan ainoastaan diskreetissä pisteistössä. Tämä on
tyypillistä kaikille numeerisille DY:n ratkaisumenetelmille.

<hr class="theorem-block-end" style="border:0; border-top:2px solid #00695c; margin-top:0.6em; margin-bottom:1.2em;">

<hr class="theorem-block-start" style="border:0; border-top:4px solid #00695c; margin-top:1.2em; margin-bottom:0.6em;">

#### Remark

Simulaatioissa voi joutua $\Delta t$:tä pienentämään pienemmäksi kuin
näytön piirtotaajuus. Tällöin jokaista ruutua kohden joudutaan
laskemaan useampia askelia.

#### Example 60

Tarkastellaan
kappaleen liikettä 2d-koordinaatistossa ajan funktiona. Asetetaan
kappaleen massaksi $m=1$ ja tiedetään, että kiihtyvyys
$\mathbf{a}(t)=\begin{pmatrix}-\cos t\\ -\sin t\end{pmatrix}$ ja alkunopeus, hetkellä $t=0$,
$\mathbf{v_0} = \begin{pmatrix}1 \\ 0\end{pmatrix}$.  
Tilannetta voidaan mallintaa DY:llä
$$\mathbf{v}'(t) = \begin{pmatrix}-\cos t\\ -\sin t\end{pmatrix}\ .$$ Valitaan
$\Delta t = 0.1$. Tällöin nopeuden kehitystä voidaan approksimoida
laskemalla
$$\mathbf{v}_{n+1} = \mathbf{v}_n + 0.1 \begin{pmatrix}-\cos t_n\\ -\sin t_n\end{pmatrix}\ ,$$
jossa $t_n = n \Delta t$.

<hr class="theorem-block-end" style="border:0; border-top:2px solid #ef6c00; margin-top:0.6em; margin-bottom:1.2em;">

<hr class="code-block-start" style="border:0; border-top:4px solid #455a64; margin-top:1.2em; margin-bottom:0.6em;">

```python
%reset -f

# Source file: examples/eulerexample01.py

import numpy as np 
from matplotlib import pyplot as plt 
from matplotlib import cm

def f(t, v):
    return np.array([-np.cos(t), -np.sin(t)])
```

```python
Dt=0.1
t0=0
v0=np.array([1, 0])
```

```python
v=v0
t=t0
N_steps=40
t_list=[t0]
v_list=[v0]
```

```python
for i in range(1, N_steps+1):
    v=v+Dt*f(t, v)
    v_list.append(v)
    t+=Dt
    t_list.append(t)
```

```python
v_list=np.array(v_list)
t_list=np.array(t_list)

for v_x, v_y, c in zip(v_list[:, 0], v_list[:, 1], cm.brg(t_list/max(t_list))):
    plt.arrow(0, 0, v_x, v_y, color=c, width=0.01)
for t, x in zip(t_list[0::5], v_list[0::5]):
    plt.annotate(f't={t:.1f}', x)
plt.show()
```

#### Definition

Olkoon $\mathbf{a} \in \mathbb{R}^n$ ja $\mathbf{b} \in \mathbb{R}^m$. Tällöin
$$\begin{pmatrix}\mathbf{a}\\ \mathbf{b}\end{pmatrix} = \begin{pmatrix}a_1\\a_2\\ \vdots \\ a_n \\ b_1\\ b_2 \\ \vdots \\ b_m\end{pmatrix} \ .$$
Vastaavasti toimitaan jos vektoreita on useampia.

<hr class="theorem-block-end" style="border:0; border-top:2px solid #1565c0; margin-top:0.6em; margin-bottom:1.2em;">

<hr class="theorem-block-start" style="border:0; border-top:4px solid #00695c; margin-top:1.2em; margin-bottom:0.6em;">

#### Remark

Edellä esitetty muoto (1)
on ensimmäisen asteen DY. Menetelmillä voidaan kuitenkin ratkaista
myös korkeamman kertaluvun yhtälöitä. Asetetaan 

$$
\begin{aligned}
            \mathbf{y}_0 & = \mathbf{y}\\
            \mathbf{y}_1 & = \mathbf{y}'\\
            \mathbf{y}_2 & = \mathbf{y}''\ \textnormal{ jne.}
\end{aligned}
$$

 Kirjoitetaan DY muotoon
$$\begin{pmatrix}\mathbf{y}'_0 \\ \mathbf{y}'_1 \\ \mathbf{y}'_2 \\ \vdots \\ \mathbf{y}'_n\end{pmatrix} = \mathbf{f}(t,\, \begin{pmatrix}\mathbf{y}_0(t)\\ \mathbf{y}_1(t) \\ \mathbf{y}_2(t)\\ \vdots \\ \mathbf{y}_n(t)\end{pmatrix})\ .$$

#### Example 61

Lasketaan esimerkin 60
tilanteessa myös kappaleen paikka.  
Fysiikasta tiedetään, että $\mathbf{a}=\mathbf{v}'$ ja edelleen $\mathbf{v}=\mathbf{x}'$.
Kirjoitetaan nämä samaan yhtälöön 

$$
\begin{aligned}
      \begin{pmatrix}\mathbf{x}' \\ \mathbf{v}'\end{pmatrix} & = \begin{pmatrix}\mathbf{v} \\ \mathbf{a}\end{pmatrix}
\end{aligned}
$$

 Kirjoitetaan tähän annettu $\mathbf{a}$, jolloin saadaan
$$\begin{pmatrix}\mathbf{x}' \\ \mathbf{v}'\end{pmatrix} = \begin{pmatrix} \mathbf{v} \\ -\cos t\\ -\sin t\end{pmatrix}\ .$$
Jotta ratkaisu olisi yksikäsitteinen, tarvitaan
$\mathbf{x}(0)= \mathbf{x}_0=\begin{pmatrix}1\\0\end{pmatrix}$.

#### Example 61

Ratkaistavana on siis

> *Tässä kaikki termit on kirjoitettu erilleen. Merkintä on epäselvä,
> joten jatkossa käytetään lyhyempää merkintää.*

$$\begin{pmatrix}{x}'_x \\ {x}'_y \\ {v}'_x \\ {v}'_y\end{pmatrix} = \begin{pmatrix} v_x \\ v_y \\ -\cos t\\ -\sin t\end{pmatrix}\ ,$$
joka on 1-kl DY. Tämä voidaan ratkaista Eulerin menetelmällä.
Lasketaan
$$\begin{pmatrix}\mathbf{x}_{n+1} \\ \mathbf{v}_{n+1}\end{pmatrix} = \begin{pmatrix}\mathbf{x}_{n} \\ \mathbf{v}_{n}\end{pmatrix} + \Delta t \begin{pmatrix} \mathbf{v}_n \\ -\cos t\\ -\sin t\end{pmatrix}\ .$$

<hr class="theorem-block-end" style="border:0; border-top:2px solid #ef6c00; margin-top:0.6em; margin-bottom:1.2em;">

<hr class="code-block-start" style="border:0; border-top:4px solid #455a64; margin-top:1.2em; margin-bottom:0.6em;">

```python
%reset -f

# Source file: examples/eulerexample02.py

import numpy as np 
from matplotlib import pyplot as plt 
from matplotlib import cm

def f(t, z):
    return np.array([z[2], z[3], 
                     -np.cos(t), -np.sin(t)])
```

```python
Dt=0.1
t0=0
z0=np.array([1, 0, 1, 0])
```

```python
z=z0
t=t0
N_steps=40
t_list=[t0]
z_list=[z0]
```

```python
for i in range(1, N_steps+1):
    z=z+Dt*f(t, z)
    z_list.append(z)
    t=t+Dt
    t_list.append(t)
```

```python
z_list=np.array(z_list)
x_list=z_list[:, 0:2]
t_list=np.array(t_list)

plt.scatter(x_list[:, 0], x_list[:, 1], 
            c=cm.brg(t_list/max(t_list)))
for t, x in zip(t_list[0::5], x_list[0::5]):
    plt.annotate(f't={t:.1f}', x)
plt.show()
```

#### Example 62

-   2.-kl yhtälö, valitaan $v(t)=x\textnormal{'}(t)$

-   Ratkaistaan 

$$
\begin{aligned}
                    {v}'(t) & = {x}''(t) = 8 \sin 4t - 8 {x}'(t) -16 x(t) \\
                    & = 8 \sin 4t - 8 v(t) -16 x(t)\ .
    \end{aligned}
$$

-   Kirjoitetaan muokattu DY vektorimuotoon
    $$\begin{pmatrix}{x}'(t) \\ {v}'(t)\end{pmatrix} = \begin{pmatrix}v(t) \\ 8 \sin 4t - 8 v(t) -16 x(t)\end{pmatrix} \ .$$

-   Tästä saadaan aika-askel
    $$\begin{pmatrix}x_{n+1} \\ v_{n+1}\end{pmatrix} = \begin{pmatrix}x_{n} \\ v_{n}\end{pmatrix} + \Delta t \begin{pmatrix}v_n \\ 8 \sin 4t_n - 8 v_n -16 x_n\end{pmatrix} \ .$$

<hr class="theorem-block-end" style="border:0; border-top:2px solid #ef6c00; margin-top:0.6em; margin-bottom:1.2em;">

<hr class="code-block-start" style="border:0; border-top:4px solid #455a64; margin-top:1.2em; margin-bottom:0.6em;">

```python
%reset -f

# Source file: examples/eulerexample03.py

import numpy as np 
from matplotlib import pyplot as plt 
from matplotlib import cm

def f(t, z):
    x=z[0]
    v=z[1]
    return np.array([v, 8*np.sin(4*t)-8*v-16*x])
```

```python
def solution(t):
    return 0.25*np.exp(-4*t)+t*np.exp(-4*t)-0.25*np.cos(4*t)
```

```python
Dt=0.05
t0=0
z0=np.array([0, 0])
```

```python
z=z0
t=t0
N_steps=100
t_list=[0]
z_list=[z0]
solution_list=[0]
```

```python
for i in range(1, N_steps+1):
    z=z+Dt*f(t, z)
    z_list.append(z)
    t=t+Dt
    t_list.append(t)
    solution_list.append(solution(t))
```

```python
z_list=np.array(z_list)
t_list=np.array(t_list)
solution_list=np.array(solution_list)

plt.plot(t_list, z_list[:, 0], label='Numeerinen')
plt.plot(t_list, solution_list, label='Analyyttinen')
plt.legend()
plt.show()
```
