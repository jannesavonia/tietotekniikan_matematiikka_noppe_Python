# RSA-salaus

## Algoritmi

<hr class="theorem-block-start" style="border:0; border-top:4px solid #0277bd; margin-top:1.2em; margin-bottom:0.6em;">

#### Algorithm: RSA avaimet

Vastaanottaja generoi

1.  Valitse $p,q \in \mathbb{P}$ ja laske $n=pq$

2.  Laske $\varphi(pq)=(p-1)(q-1)$

3.  Valitse "pieni eksponentti" $e$ siten, että

    -   $\operatorname{syt}\left(e,\,\varphi(pq)\right)=1$

    -   Varma valinta on pienehkö alkuluku

    -   Yleisesti käytetään lukua $65537$

4.  Valitse $d$ siten, että

    -   $de \equiv 1 \pmod{\varphi(pq)}$

    -   "Modulaarinen käänteisluku"

5.  Julkinen avain $K_{pub}=(n,\, e)$

6.  Yksityinen avain $K_{priv}=(n,\, d)$

<hr class="theorem-block-end" style="border:0; border-top:2px solid #0277bd; margin-top:0.6em; margin-bottom:1.2em;">

<hr class="theorem-block-start" style="border:0; border-top:4px solid #0277bd; margin-top:1.2em; margin-bottom:0.6em;">

#### Algorithm: Viestin salaus

1.  Vastaanottaja lähettää julkisen avaimen $K_{pub}=(n,\, e)$ viestin
    lähettäjälle

2.  Salattava viesti $M<n$

    -   Voidaan jakaa osiin

3.  Lasketaan $$C \equiv M^e \pmod{n}$$

<hr class="theorem-block-end" style="border:0; border-top:2px solid #0277bd; margin-top:0.6em; margin-bottom:1.2em;">

<hr class="theorem-block-start" style="border:0; border-top:4px solid #0277bd; margin-top:1.2em; margin-bottom:0.6em;">

#### Algorithm: Viestin purkaminen

1.  Vastaanottajalla on yksityinen avain $K_{priv}=(n,\, d)$

2.  Vastaanotettu viesti $C<n$

3.  Lasketaan $$M \equiv C^d \pmod{n}$$

<hr class="theorem-block-end" style="border:0; border-top:2px solid #0277bd; margin-top:0.6em; margin-bottom:1.2em;">

### $d$:n laskeminen

Pitää löytää $d$ siten, että 

$$
\begin{aligned}
de & \equiv 1 \pmod{ \varphi(pq)} \\
\Rightarrow de -1 & \equiv 0 \pmod{ \varphi(pq)} \\
\Rightarrow de -1 & = (-y) \varphi(pq)\ , \ \textnormal{ jollekin } y\in \mathbb{Z} \\
\Rightarrow de + y\varphi(pq) & = 1 = \operatorname{syt}\left(e,\,\varphi(pq)\right) \end{aligned}
$$

<hr class="theorem-block-start" style="border:0; border-top:4px solid #0277bd; margin-top:1.2em; margin-bottom:0.6em;">

#### Algorithm

1.  Merkitään $m=\varphi(pq)$

2.  Lasketaan $(s, x, y) = \overline{\rm syt}(e,\,m)$

    -   $\overline{\rm syt}(e,\,m)$ lasketaan laajennetulla Eukleideen
        algoritmilla ja se palauttaa myös kertoimet $x$ ja $y$

    -   Jos kaikki on mennyt oikein, niin $s=1$

3.  $d \equiv x \pmod{m}$

    -   Pythonilla <span class="nodecor">`d=x%m`</span>

<hr class="theorem-block-end" style="border:0; border-top:2px solid #0277bd; margin-top:0.6em; margin-bottom:1.2em;">

### Miksi RSA toimii?

-   Ei ole tunnettua algoritmia, joilla suurten lukujen jako tekijöihin
    onnistuisi

    -   $n$:stä ei voida laskea $p$:tä tai $q$:ta

    -   $e$:n laskeminen ei onnistu (ainakaan nykytiedon valossa)

    -   Murtaminen ei onnistu

-   $C^d \equiv (M^e)^d \equiv M^{ed} \equiv M \pmod{n}$
