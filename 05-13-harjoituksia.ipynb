## Harjoituksia

### Harjoituksia

-   **20.** Olkoon $\mathbf{x}=\begin{pmatrix}1 & -1 & 0\end{pmatrix}$,
    $\mathbf{y}=\begin{pmatrix}3 & -2 & 0\end{pmatrix}$, $\mathbf{z}=\begin{pmatrix}1& 2\end{pmatrix}$ ja
    $\mathbf{w}=\begin{pmatrix}-2 & -1\end{pmatrix}$. Mikäli laskutoimitukset on määritelty,
    niin laske

    1.  $\mathbf{x}+\mathbf{y}$,   $\mathbf{x}-\mathbf{y}$,   $\mathbf{x}+\mathbf{w}$,   $\mathbf{z}+\mathbf{w}$
         ja $\mathbf{y}-\mathbf{y}$.

    2.  $\lVert \mathbf{x}+\mathbf{y} \rVert$,   $\lVert \mathbf{x}-\mathbf{y} \rVert$,
          $\lVert \mathbf{x} \rVert+\lVert \mathbf{w} \rVert$,   $\lVert \mathbf{z} \rVert+\mathbf{w}$,
          $\lVert \mathbf{z} \rVert+\lVert \mathbf{w} \rVert$  ja $\lVert \mathbf{y} \rVert\mathbf{y}$.

-   **21.** Olkoon vektorit kuten edellä. Laske kunkin vektorin kanssa
    samansuuntainen yksikkövektori.

-   **22.** Jos $\hat{\mathbf{x}}=\hat{\mathbf{y}}$ ja $\lVert \mathbf{x} \rVert=\lVert \mathbf{y} \rVert$, niin onko
    $\mathbf{x}=\mathbf{y}$?

-   **M16.** Lisää `Vect`-luokkaan funktio `unit()`, joka palauttaa
    vektorin kanssa samansuuntaisen yksikkövektorin. Mikäli vektori on
    $0$-vektori, eli sen pituus on $0$, niin heitetään poikkeus.
