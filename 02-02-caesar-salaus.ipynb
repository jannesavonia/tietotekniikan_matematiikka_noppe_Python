# Salaukset

## Caesar-salaus

### Caesar-salaus

-   [Caesar-salakirjoitus](https://fi.wikipedia.org/wiki/Caesarin_salakirjoitus)

-   Jokainen kirjain korvataan $K$ kirjaimen päässä olevalla kirjaimella

-   Aakkosten viimeisen kirjaimen jälkeen tulee aakkosten ensimmäinen
    kirjain

-   Välimerkit voidaan välittää esimerkiksi sellaisenaan (tai poistaa)

    > *Tällä kurssille ne välitetään muuttumattomina*

-   Käytetään englannin isoja aakkosia (`"A" …"Z"`)

-   Salaaminen useampaan kertaan ei tuo lisäturvaa

### Esimerkki

1.  Poistetaan "ääkköset"

2.  Muutetaan kirjaimet isoiksi

3.  $M$=`"ET OSAA MURTAA TATA!"`

4.  Valitaan avaimeksi $K=16$

    -   Siirretään kirjaimia 16 positiota eteenpäin

    -   Esimerkiksi `"A"`$\mapsto$`"Q"`

5.  $C$=`"UJ EIQQ CKHJQQ JQJQ!"`

6.  Vastaanottaja vastaanottaa $C$:n

7.  Koska vastaanottaja tietää avaimen $K$, hän pystyy purkamaan
    salauksen

    -   Siirretään kirjaimia 16 positiota taaksepäin

    -   Esimerkiksi `"Q"`$\mapsto$`"A"`

8.  $M_{received}$=`"ET OSAA MURTAA TATA!"`

### Murtaminen

-   Vähän mahdollisia avaimia, joten kokeileminen on varteenotettava
    vaihtoehto

#### Example 1

-   Vastaanotettu viesti  
    $C$=`"WFFWPEVZ ZDLLGLDT AFCVLL DLWLFVDPY?"`

-   Tiedetään, että se on salattu Caesar-salakirjoituksella

-   Kokeillaan kaikki avaimet läpi:

``` text
K=0 : WFFWPEVZ ZDLLGLDT AFCVLL DLWLFVDPY?
....
K=10 : MVVMFULP PTBBWBTJ QVSLBB TBMBVLTFO?
K=11 : LUULETKO OSAAVASI PURKAA SALAUKSEN?
K=12 : KTTKDSJN NRZZUZRH OTQJZZ RZKZTJRDM?
....
K=25 : XGGXQFWA AEMMHMEU BGDWMM EMXMGWEQZ?  
```

<hr class="theorem-block-end" style="border:0; border-top:2px solid #ef6c00; margin-top:0.6em; margin-bottom:1.2em;">
