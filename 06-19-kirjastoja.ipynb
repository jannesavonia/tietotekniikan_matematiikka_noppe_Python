```python
%reset -f

# Source file: examples/numpyexample01.py

import numpy as np
```

```python
#Create vector
x=np.array([1.0, 2.0, -1.1, 1])
print(f'x={x}')
```

```python
#Create matrix
A=np.array([[1, 4, 1.1, 2],[-0.6, 1, 2, 3]])
print(f'A={A}')
```

```python
#Multiplication
Ax=A@x
print(f'Ax={Ax}')
```

```python
#Save matrix
np.save('Ax.npy', Ax)
```

```python
del Ax
```

```python
#Load matrix
Ax=np.load('Ax.npy')
print(f'Ax={Ax}')
```

```python
%reset -f

# Source file: examples/numpyexample02.py

import numpy as np 
from matplotlib import pyplot as plt 

x=np.array([1.0, 3])
y=np.array([3.2, 5.1])
```

```python
#Compute dot product
xy=x@y
print(f'{x} @ {y} = {xy}')
```

```python
O=np.array([0., 0.])

plt.figure()
plt.axis('equal')
x_list, y_list = zip(O, x)
plt.plot(x_list, y_list, '-*')
x_list, y_list = zip(O, y)
plt.plot(x_list, y_list, ':')
plt.show()
```

```python
x_minus_y=x-y
```

```python
def arrow(x, y, dx, dy, color, label):
    return plt.arrow(x, y, dx, dy, 
                     width=0.001, head_width=0.1,
                     head_length=0.15, 
                     color=color, overhang=0.6, 
                     length_includes_head=True), label
```

```python
plt.figure()
plt.title('x-y')
plt.axis('equal')
a1, l1=arrow(0, 0, x[0], x[1],
             color='r', label='x')
a2, l2=arrow(0, 0, y[0], y[1], 
             color='k', label='y')
a3, l3=arrow(y[0], y[1], x_minus_y[0], x_minus_y[1],
             color='b', label='x-y')
plt.legend([a1, a2, a3], [l1, l2, l3])
plt.show()
```

```python
minus_y=-y
```

```python
plt.figure()
plt.title('x-y')
plt.axis('equal')
a1, l1=arrow(0, 0, x[0], x[1],
             color='r', label='x')
a2, l2=arrow(x[0], x[1], minus_y[0], minus_y[1], 
             color='k', label='-y')
a3, l3=arrow(0, 0, x_minus_y[0], x_minus_y[1], 
             color='b', label='x-y')
plt.legend([a1, a2, a3], [l1, l2, l3])
plt.show()
```

```python
%reset -f

# Source file: examples/numpyexample03.py

import numpy as np 
from matplotlib import pyplot as plt 

#Insert points into list (forms matrix)
x=np.array([[0, 0], 
            [2, 1], 
            [2, 2], 
            [1, 2], 
            [0, 0],
            [1, 1],
            [2, 2]], dtype=np.double).transpose()
print(x)
```

```python
#Create linear mapping (matrix)
A=np.array([[2, 0],
            [0, -1]])
```

```python
Ax=A@x
```

```python
Ax_1, Ax_2=Ax
x_1, x_2=x
```

```python
plt.plot(x_1, x_2, '-*')
plt.plot(Ax_1, Ax_2, '-*')
plt.show()
```
