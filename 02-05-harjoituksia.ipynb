# Salaukset

## Harjoituksia

### Harjoituksia

-   **M3.** Toteuta substituutiosalain ja purku.

-   **M4.** Kirjoita ohjelma, joka purkaa substituutiosalauksella
    salatun viestin. Ohjelma tulostaa puretun viestin. Salattu viesti
    sisältää isoja englannin aakkosia `A …Z`. Välilyönnit, numerot ja
    erikoismerkit ovat koodaamattomina.
