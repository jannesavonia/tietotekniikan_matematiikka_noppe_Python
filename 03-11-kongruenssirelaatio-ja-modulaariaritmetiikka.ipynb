#### Definition

Olkoon $a,\,b \in \mathbb{Z}$. Tällöin $a$ ja $b$ **ovat kongruentteja
modulo** $n$, merkitään $$a \equiv b \pmod{n}\ ,$$ jos
$n \vert (a-b)$.

#### Example 13

-   $7 \equiv 1 \pmod{3}$, koska $3 \vert (7-1)$

    -   Vaihtoehtoisesti löytyy $m \in \mathbb{Z}$ s.e $7=m\cdot3+1$

    -   $m=2$

-   $2 \equiv 14 \pmod{12}$, koska $12 \vert (2-14)$

    -   Vaihtoehtoisesti löytyy $m \in \mathbb{Z}$ s.e $2=m\cdot12+14$

    -   $m=-1$

-   $7 \not \equiv 1 \pmod{4}$, koska $4 \nmid (7-1)$

#### Theorem

1.  $a \equiv a \pmod{n}$

2.  $a \equiv b \pmod{n} \Leftrightarrow b \equiv a \pmod{n}$

3.  Jos $a \equiv b \pmod{n}$ ja $b \equiv c \pmod{n}$, niin
    $a \equiv c \pmod{n}$

4.  Jos $a \equiv b \pmod{n}$ ja $c \equiv d \pmod{n}$, niin

    1.  $a+c \equiv b+d \pmod{n}$

    2.  $ac \equiv bd \pmod{n}$

5.  Jos $a \equiv b \pmod{n}$, niin

    1.  $a+c \equiv b+c \pmod{n}$

    2.  $ac \equiv bc \pmod{n}$

6.  Jos $a \equiv b \pmod{n}$, niin $a^k \equiv b^k \pmod{n}$, kun
    $k \in \mathbb{N}$

#### Remark

Edelliset laskusäännöt sanovat lyhyesti, että kun $n$ ei muutu, niin
laskutoimitukset toimivat melkein kuten kokonaisluvuilla laskettaessa.

#### Example 14

Lasketaan $$9 + 4 \equiv 13 \equiv 1 \pmod{12} \ .$$

#### Remark

Kuten edellisestä esimerkistä huomattiin, niin esitysmuoto ei ole
yksikäsitteinen.

-   Jatkossa tehtävissä oletetaan, että lukuarvot ovat välillä
    $0 \dots n-1$

    -   Useissa ohjelmointikielissä jakojäännös `k%n` on aina välillä
        $0\dots n-1$

-   Paperilla laskettaessa "lisää tai vähennä $n$, kunnes saat halutun
    esitysmuodon"

#### Example 15

-   $6+7 \equiv 13 \equiv 1 \pmod{4}$

-   $6-7 \equiv 6 + (-7) \equiv -1 \equiv 3 \pmod{4}$

-   $6 \cdot 7 \equiv 42 \equiv 2 \pmod{4}$

-   $2\cdot 2 -5 \equiv 4 + (-5) \equiv -1 \equiv 2 \pmod{3}$

-   $2\cdot (2 -5) \equiv 2 \cdot (-3) \equiv -6 \equiv 0 \pmod{3}$

#### Example 16

-   Lasketaan $16+19+55 \pmod{5}$:

    -   $16 \equiv 1 \pmod{5}$ , $19 \equiv -1 \pmod{5}$ ja
        $55 \equiv 0 \pmod{5}$

    -   $16+19+55 \equiv 1-1+0 \equiv 0 \pmod{5}$

-   Lasketaan $16 \cdot 19 \cdot 55 \pmod{5}$:

    -   $16 \cdot 19 \cdot 55 \equiv 1 \cdot(-1) \cdot 0 \equiv 0 \pmod{5}$

-   Lasketaan $91^{131} \pmod{10}$:

    -   $91 \equiv 1 \pmod{10}$

    -   $91^{131} \equiv 1^{131} \equiv 1\pmod{10}$
