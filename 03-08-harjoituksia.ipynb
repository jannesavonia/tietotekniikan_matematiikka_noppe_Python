### Harjoituksia

-   **M7.** Pieniä alkulukuja voidaan etsiä esimerkiksi [Eratostheneen
    seulan](https://fi.wikipedia.org/wiki/Eratostheneen_seula) avulla.
    Toteuta funktio `sieve(N)`, joka palauttaa joukon (`set`-tyyppiä)
    joka sisältää alkuluvut, jotka ovat enintään `N`. Testaa funktion
    toiminta vähintään lukuun `N=1000000` saakka. Ajoaika ei saa ylittää
    $10 \min$, kun `N=1000000`.  
    Testiaineistoa voi ladata esimerkiksi [Math is
    Fun-sivustolta](https://www.mathsisfun.com/numbers/prime-number-lists.html).
