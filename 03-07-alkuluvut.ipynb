## Alkuluvut

#### Definition

Luonnollinen luku $a$ on kokonaisluvun $b$ **tekijä**, jos
$a \vert b$.

#### Example 8

-   $8$ on luvun $24$ tekijä

-   $7$ ei ole luvun $24$ tekijä

-   $3$ on luvun $24$ tekijä

#### Definition

Olkoon luonnollinen luku $p>1$. $p$ on **alkuluku**, jos sen ainoat
positiiviset tekijät ovat $1$ ja $p$. Tällöin merkitään
$p \in \mathbb{P}$.

<hr class="theorem-block-end" style="border:0; border-top:2px solid #1565c0; margin-top:0.6em; margin-bottom:1.2em;">

<hr class="theorem-block-start" style="border:0; border-top:4px solid #6a1b9a; margin-top:1.2em; margin-bottom:0.6em;">

#### Theorem: Aritmetiikan peruslause

Jokainen luonnollinen luku $n>1$ voidaan esittää yksikäsitteisesti
alkulukujen äärellisenä tulona.

<hr class="theorem-block-end" style="border:0; border-top:2px solid #6a1b9a; margin-top:0.6em; margin-bottom:1.2em;">

<hr class="theorem-block-start" style="border:0; border-top:4px solid #6a1b9a; margin-top:1.2em; margin-bottom:0.6em;">

#### Theorem: Alkulukujen äärettömyys

Alkulukuja on ääretön määrä.

<hr class="theorem-block-end" style="border:0; border-top:2px solid #6a1b9a; margin-top:0.6em; margin-bottom:1.2em;">

### Todistus – (ei pakollinen osa kurssia)

-   Todistus perustuu ns. antiteesiin, joka todistetaan vääräksi

-   Antiteesi on valittu siten, että sen kumoaminen todistaa
    alkuperäisen väitteen

#### Proof

-   *Antiteesi*: Alkulukuja on äärellinen määrä

-   Tällöin alkuluvut voidaan luetella $p_1,\, p_2,\, \dots p_n$

-   Olkoon $P=p_1p_2\dots p_n$

-   Jokainen alkuluku jakaa $P$:n, eli $p_i \vert P$ kaikilla
    $i=1\dots n$

-   Pienin mahdollinen alkuluku on $2$, eli jokainen $p_i>1$

-   Mikään alkuluku $p_i$ ei voi jakaa lukua $P+1$, eli
    $p_i \nmid P+1$ kaikilla $i=1 \dots n$

-   $P+1$ on siis alkuluku ja suurempi kuin mikään luetelluista
    alkuluvusta

-   Tämä aiheuttaa ristiriidan, joten antiteesi oli väärä
