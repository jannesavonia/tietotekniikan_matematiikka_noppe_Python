# Salaukset

## Merkinnät

### Yleiset merkinnät

-   $M$ = Selkoteksti

-   $C$ = Salattu viesti

-   $E_{K_1}$ = Salausmenetelmä salausavaimella $K_1$

-   $D_{K_2}$ = Purkumenetelmä purkuavaimella $K_2$

   
Salaus: 

$$
\begin{aligned}
  C & = E_{K_1}(M) = \texttt{E(K1, M)}  \end{aligned}
$$

 Purku: 

$$
\begin{aligned}
  M & = D_{K_2}(C) = \texttt{D(K2, C)}\end{aligned}
$$
