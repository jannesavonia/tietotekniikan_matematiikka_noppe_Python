## Eukleideen algoritmi

### Eukleideen algoritmi

-   Suurimmalle yhteiselle tekijälle pätee
    $$\operatorname{syt}\left(a,\,b\right) = \operatorname{syt}\left(\lvert a \rvert,\,\lvert b \rvert\right)\ \textnormal{ ja }\ \operatorname{syt}\left(a,\,b\right) = \operatorname{syt}\left(b,\,a\right)$$

-   Tämän vuoksi algoritmi määritellään vain kokonaisluvuille
    $a\ge b>0$.

<hr class="theorem-block-start" style="border:0; border-top:4px solid #0277bd; margin-top:1.2em; margin-bottom:0.6em;">

#### Algorithm: Eukleideen algoritmi

Olkoon $a\ge b>0$

1.  Asetetaan $\hat{a}=a$ ja $\hat{b}=b$

2.  Ratkaise $r$ jakoyhtälöstä $\hat{a} = q\hat{b}+r$

    -   $r$ on jakojäännös

    -   Monissa ohjelmointikielissä
        <span class="nodecor">`r=a%b`</span>

3.  Jos $r=0$, niin tällöin $\operatorname{syt}\left(a,\,b\right)=\hat{b}$

4.  Asetetaan $\hat{a} = \hat{b}$ ja $\hat{b} = r$

5.  Jatka kohdasta (2)

#### Example 6

Lasketaan $\operatorname{syt}\left(12378,\,3054\right)$: 

$$
\begin{aligned}
\hat{a}=12378& \hat{b}=3054 \Rightarrow r=162 \\
\hat{a}=3054& \hat{b}=162 \Rightarrow r=138 \\
\hat{a}=162& \hat{b}=138 \Rightarrow r=24 \\
\hat{a}=138& \hat{b}=24 \Rightarrow r=18 \\
\hat{a}=24& \hat{b}=18 \Rightarrow r=6 \\
\hat{a}=18& \underbrace{\hat{b}=6}_{=\operatorname{syt}\left(12378,\,3054\right)} \Rightarrow r=0 \\
\end{aligned}
$$
